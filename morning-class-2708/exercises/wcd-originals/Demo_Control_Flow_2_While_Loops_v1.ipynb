# Part 02 - while Loops

````python
count = 0
while count < 5:
    print(count)
    count += 1
````

**What you should see:** `0, 1, 2, 3, 4`, each on its own line. A `while` loop keeps running as long as its condition (`count < 5`) stays `True`; `count += 1` is what eventually makes the condition `False` and ends the loop. Forgetting to update the loop variable is the classic way to create an accidental infinite loop.

````python
# # Infinite loop
# while True:
#     print("This is an infinite loop!")
````

**What you should see:** no output, this cell is entirely commented out. Uncommenting and running it would start a loop that never stops on its own, since `True` is always `True`, there's no condition to become `False`. This is why it's left commented in the notebook rather than run live: an infinite loop would need to be manually interrupted (the Jupyter "stop" button, or Kernel → Interrupt).

````python
count = 0
while count < 5:
    if count == 3:
        break  # Terminate loop when count is 3
    print(count)
    count += 1
````

**What you should see:** `0, 1, 2`, then the loop stops, `3` never prints. `break` exits the loop immediately, skipping the rest of the loop body and any remaining iterations, as soon as `count == 3`.

````python
for i in range(5):
  if i == 3:
    break
  print(i)
````

**What you should see:** `0, 1, 2`, the `for`-loop equivalent of the cell above. `break` works the same way in both `while` and `for` loops.

````python
count = 0
while count < 5:
    if count == 3:
        count += 1
        continue
    print(count)
    count += 1
````

**What you should see:** `0, 1, 2, 4`, note `3` is skipped but the loop keeps going afterward. `continue` skips the rest of the *current* iteration only, unlike `break` it doesn't end the loop. Note the `count += 1` is duplicated inside the `if` branch here, since `continue` jumps straight back to the condition check, the loop would otherwise get stuck re-testing `count == 3` forever without the `if` branch's own increment.

````python
for i in range(5):
  if i == 3:
    continue
  print(i)
````

**What you should see:** `0, 1, 2, 4`, the `for`-loop equivalent. Here `continue` is simpler to reason about, since `for` automatically advances to the next item regardless, there's no separate counter to remember to increment.

````python
count = 0
while count < 5:
    print(count)
    count += 1
else:
    print("Loop completed successfully!")
````

**What you should see:**
```
0
1
2
3
4
Loop completed successfully!
```
Just like a `for` loop, a `while` loop's `else` block runs once the loop finishes normally, i.e. once its condition becomes `False` on its own, without a `break`.

````python
while True:
    age_str = input("Enter your age: ")
    if age_str.isdigit():
        age = int(age_str)
        if age >= 0:
            break
    print("Invalid input. Please enter a non-negative integer.")
````

**Note:** this cell uses `input()` and pauses waiting for you to type a response. It only works interactively in a live Jupyter session. Enter any non-negative whole number (like `25`) to exit the loop, or try entering something invalid first (like `abc` or `-5`) to see the reprompt.

**What you should see:** the prompt `Enter your age: ` repeats with `Invalid input. Please enter a non-negative integer.` after each invalid entry (non-numeric text, or a negative number), until you enter a valid non-negative integer, at which point the loop exits via `break`.

````python
while True:
  user = input('Username: ')
  password = input('password: ')

  if (user == 'waqas') and (password == '1234'):
    print()
    print("Welcome!")
    break
  else:
    print()
    print('Incorrect username and password. Try again!')
````

**Note:** this cell also uses `input()` (twice, for username and password) and pauses waiting for you to type. It only works interactively in a live Jupyter session. The correct credentials are username `waqas`, password `1234`.

**What you should see:** `Welcome!` if you enter `waqas` / `1234`, otherwise `Incorrect username and password. Try again!` and the prompts repeat until you get it right.
