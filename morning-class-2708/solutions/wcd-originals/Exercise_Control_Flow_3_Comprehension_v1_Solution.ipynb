# **Part 03 - List and Dictionaries Comprehensions**

---

### Create a list comprehension that generates squares of numbers from 0 to 4.

```python
squares = [x**2 for x in range(5)]
squares
```

**What you should see:** `[0, 1, 4, 9, 16]`.

### Create a list comprehension that generates squares of even numbers from 0 to 9.

```python
even_squares = [x**2 for x in range(10) if x % 2 == 0]
even_squares
```

**What you should see:** `[0, 4, 16, 36, 64]`.

### Create a dictionary comprehension that maps each number from 0 to 4 to its square.

```python
square_dict = {x: x**2 for x in range(5)}
square_dict
```

**What you should see:** `{0: 0, 1: 1, 2: 4, 3: 9, 4: 16}`.

### Create a dictionary comprehension that maps each even number from 0 to 9 to its square.


```python
even_square_dict = {x: x**2 for x in range(10) if x % 2 == 0}
even_square_dict
```

**What you should see:** `{0: 0, 2: 4, 4: 16, 6: 36, 8: 64}`.

### Create a nested list comprehension that generates a 3x3 matrix where each element is the product of its row and column indices.

```python
matrix = [[i*j for j in range(3)] for i in range(3)]
matrix
```

**What you should see:** `[[0, 0, 0], [0, 1, 2], [0, 2, 4]]`, each inner list is one row, row `i`, column `j` holds `i*j`.

### Create a tuple comprehension that generates squares of numbers from 0 to 4.


```python
tuple_comp = tuple(x**2 for x in range(5))
tuple_comp
```

**What you should see:** `(0, 1, 4, 9, 16)`. Note there's technically no such thing as a "tuple comprehension" in Python, `(x**2 for x in range(5))` is a **generator expression**, and wrapping it in `tuple(...)` is what collects its values into a tuple.

### Create a set comprehension that generates squares of numbers from 0 to 4.

```python
set_comp = {x**2 for x in range(5)}
set_comp
```

**What you should see:** `{0, 1, 4, 9, 16}`.
