### Create a while loop that prints numbers from 0 to 4.

````python
count = 0
while count < 5:
    print(count)
    count += 1
````

**What you should see:** `0, 1, 2, 3, 4`, one per line.

### Modify the first while loop to terminate when the count is 3 using a `break` statement.

````python
count = 0
while count < 5:
    if count == 3:
        break
    print(count)
    count += 1
````

**What you should see:** `0, 1, 2`, then the loop stops before printing `3`.

### Create a for loop using `range(5)` and terminate it when the loop variable is 3 using a `break` statement.

````python
for i in range(5):
    if i == 3:
        break
    print(i)
````

**What you should see:** `0, 1, 2`, the `for`-loop version of the same `break` behavior.

### Modify the second while loop to skip printing when the count is 3 using a `continue` statement.

````python
count = 0
while count < 5:
    if count == 3:
        count += 1
        continue
    print(count)
    count += 1
````

**What you should see:** `0, 1, 2, 4`, `3` is skipped by `continue` but the loop keeps going.

### Modify the second for loop to skip printing when the loop variable is 3 using a `continue` statement.

````python
for i in range(5):
    if i == 3:
        continue
    print(i)
````

**What you should see:** `0, 1, 2, 4`, the `for`-loop version.

### Add an else block to the first while loop that prints `"Loop completed successfully!"` after the loop is completed.

````python
count = 0
while count < 5:
    print(count)
    count += 1
else:
    print("Loop completed successfully!")
````

**What you should see:**
```
0
1
2
3
4
Loop completed successfully!
```

### Create a while loop that prompts the user to enter their age. If the input is a non-negative integer, exit the loop; otherwise, continue prompting.

````python
while True:
    age_str = input("Enter your age: ")
    if age_str.isdigit():
        age = int(age_str)
        if age >= 0:
            break
    print("Invalid input. Please enter a non-negative integer.")
````

**Note:** this cell uses `input()` and pauses waiting for you to type. It only works interactively in a live Jupyter session. Enter any non-negative whole number to exit the loop.

**What you should see:** the reprompt `Invalid input. Please enter a non-negative integer.` for any non-numeric or negative entry, until a valid non-negative integer is entered.

### Create a while loop that prompts the user to enter a username and password. If the username is `'waqas'` and the password is `'1234'`, print `"Welcome!"` and exit the loop; otherwise, print `"Incorrect username and password. Try again!"` and continue prompting.

````python
while True:
    user = input('Username: ')
    password = input('Password: ')

    if (user == 'waqas') and (password == '1234'):
        print()
        print("Welcome!")
        break
    else:
        print()
        print('Incorrect username and password. Try again!')
````

**Note:** this cell uses `input()` twice and pauses waiting for you to type. It only works interactively in a live Jupyter session. The correct credentials are username `waqas`, password `1234`.

**What you should see:** `Welcome!` once the correct username and password are entered, otherwise `Incorrect username and password. Try again!` and the prompts repeat.
