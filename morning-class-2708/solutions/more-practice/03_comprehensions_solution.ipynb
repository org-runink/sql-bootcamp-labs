```python
# Extra practice 03 — Comprehensions. Run this once.
prices = [4.5, 12.0, 0.0, 7.25, 30.0, 0.0]
skus = ["AA-1", "bb-2", "CC-3", "dd-4"]

sales = {"north": 1200, "south": 0, "east": 830, "west": 455}
matrix = [[1, 2, 3], [4, 5, 6]]

pairs = [("ana", 88), ("bo", 42), ("cai", 95), ("dee", 61)]

print("prices:", prices)
print("sales: ", sales)
```

### Question 1

Two plain maps. -> `[5.3999999999999995, 14.399999999999999, 0.0, 8.7, 36.0, 0.0] 6`, then `['AA-1', 'BB-2', 'CC-3', 'DD-4'] 4`.

Six in, six out; four in, four out. No `if`, so nothing can be dropped.

`4.5 * 1.2` came out as `5.3999999999999995` rather than `5.4`. Neither
number is stored exactly in binary, so the product carries a little error —
the same thing worksheet 02 Q9 showed with 1.19. It is fine for display
with `f"{p:.2f}"` and it is not fine as the basis of an `==` test.

`s.upper()` produced a new string. Strings are immutable; the originals in
`skus` are untouched.

```python
with_tax = [p * 1.2 for p in prices]
print(with_tax, len(with_tax))

upper_skus = [s.upper() for s in skus]
print(upper_skus, len(upper_skus))
```

### Question 2

Two filters. -> `[4.5, 12.0, 7.25, 30.0] 4 of 6`, then `['bb-2', 'dd-4'] 2 of 4`.

The `if` at the end drops items, so the results are shorter than the
inputs. Printing the length **against the input length** is the habit worth
keeping — `4 of 6` says something that `4` does not.

`s[1].islower()` indexes the second character and asks a question about it.
It would raise `IndexError` on a sku shorter than two characters, which is
the sort of thing real data supplies eventually.

```python
nonzero = [p for p in prices if p > 0]
print(nonzero, len(nonzero), "of", len(prices))

dashed = [s for s in skus if s[1].islower()]
print(dashed, len(dashed), "of", len(skus))
```

### Question 3

The map form. -> `[4.5, 10, 0.0, 7.25, 10, 0.0] 6`, then `True`.

Same length as the input, because every item became something. Compare
with Q2, where the same kind of test made the list shorter.

Note the mixed types: `10` is an int and the rest are floats, because the
cap was written as `10` rather than `10.0`. Python does not mind; a
stricter consumer downstream might.

```python
capped = [10 if p > 10 else p for p in prices]
print(capped, len(capped))
print(len(capped) == len(prices))
```

### Question 4

`any` and `all`. -> `True`, `False`, `True`, `True`.

One line, one boolean, no loop and no flag variable. This is the shape for
a validation check — *is anything wrong* / *is everything fine* — and it
replaces five lines of accumulator every time.

The brackets are missing on purpose: `any(p == 0 for p in prices)` is a
**generator expression**, which stops as soon as it finds a `True`. Write
`any([...])` with brackets and you build the whole list first, then check
it. Same answer, more work.

`any` on an empty collection is `False` and `all` on an empty collection is
**`True`** — "every one of no items passed". Surprising, occasionally
useful, and worth remembering before you use `all()` to gate a pipeline
that might have received nothing.

```python
print(any(p == 0 for p in prices))
print(all(p > 0 for p in prices))
print(all("-" in s for s in skus))
print(any(v == 0 for v in sales.values()))
```

### Question 5

Dictionary in, list out. -> `['east', 'north']`, then `['north: 1200', 'south: 0', 'east: 830', 'west: 455']`.

`sorted()` wrapped round a comprehension gives a predictable order, which
matters the moment the output is going into a report rather than another
variable.

`.items()` with two loop variables is the same unpacking as everywhere
else. `report` keeps the dictionary's own insertion order — dictionaries
have preserved that since Python 3.7 — but do not rely on it meaning
anything; it is the order the keys were added, not any kind of ranking.

```python
busy = sorted([k for k, v in sales.items() if v > 500])
print(busy)

report = [f"{k}: {v}" for k, v in sales.items()]
print(report)
```

### Question 6

Three dict comprehensions. -> `{'ana': 88, 'bo': 42, 'cai': 95, 'dee': 61} 4`; `{'ana': 88, 'cai': 95, 'dee': 61} 3`; `{'ana': 'pass', 'bo': 'fail', 'cai': 'pass', 'dee': 'pass'} 4`.

The filter version drops `bo` and comes out at 3; the map version keeps
everyone and labels them, staying at 4. That is worksheet 03 Q5 again, in
dictionary form.

`banded` has a conditional expression in the **value** position:
`{name: ("pass" if … else "fail") for …}`. The colon separates key from
value, and the `if`/`else` belongs entirely to the value — the brackets
would be redundant but they do no harm while you are learning where the
pieces go.

All four names are distinct, so nothing was overwritten. Worksheet 03 Q7
is what happens when they are not.

```python
scores = {name: score for name, score in pairs}
print(scores, len(scores))

passed = {name: score for name, score in pairs if score >= 60}
print(passed, len(passed))

banded = {name: "pass" if score >= 60 else "fail" for name, score in pairs}
print(banded, len(banded))
```

### Question 7

Nested comprehensions. -> `[1, 2, 3, 4, 5, 6]`, then `[[2, 4, 6], [8, 10, 12]]`.

These look similar and build different shapes. `flat` has **two `for`
clauses in one comprehension** and produces one flat list. `doubled_rows`
has **a comprehension inside another one's expression slot** and produces a
list of lists, one per row.

The giveaway is the inner brackets. If they are there, you get nesting
out; if they are not, you get flattening.

The clause order in `flat` — `for row in matrix for n in row` — reads outer
then inner, top to bottom as you would write the loops. Reverse them and
it is a `NameError`.

```python
flat = [n for row in matrix for n in row]
print(flat)

doubled_rows = [[n * 2 for n in row] for row in matrix]
print(doubled_rows)
```

### Question 8

One line too far. -> both versions print `['ana: B', 'cai: A', 'dee: C']`, then `True`.

Identical output. The comprehension is 2 lines and the loop is 12, and the
loop is the one you want to find at 3am — the grade boundaries are on
separate lines where you can read them, and the `continue` says out loud
that a row is being dropped.

The chained `'A' if … else 'B' if … else 'C' if … else 'F'` is legal and
nests to the right, so it reads as a fall-through chain. It also has no
names for its intermediate steps and no room for a comment.

And look at what neither version tells you: **`F` can never happen.** The
filter drops anyone under 50 and the `C` band catches 60 and above, so only
a score of 50–59 could produce an `F`, and there isn't one in this data.
That is worksheet 01 Q5's unreachable branch, hiding in a one-liner where
it is even harder to spot.

Rule of thumb: one `for` and at most one `if` belongs in a comprehension.
Beyond that, write the loop.

```python
one_liner = [f"{n}: {'A' if s >= 90 else 'B' if s >= 80 else 'C' if s >= 60 else 'F'}"
             for n, s in pairs if s >= 50]
print(one_liner)

as_loop = []
for name, score in pairs:
    if score < 50:
        continue
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 60:
        grade = "C"
    else:
        grade = "F"
    as_loop.append(f"{name}: {grade}")

print(as_loop)
print(one_liner == as_loop)
```

### Question 9

A set of tuples, then a set of lists. -> the four `(name, score)` tuples in a set, then `TypeError: unhashable type: 'list'`.

**The set will print in a different order for you.** Compare contents, not
arrangement.

Tuples are hashable so a set can hold them; lists are not, so it cannot.
That is the same rule as worksheet 03 Q11's dictionary keys, and for the
same reason — a set finds its items by hash, exactly as a dictionary finds
its keys, so anything that can change after it goes in is refused.

`{tuple(row) for row in matrix}` works. Whether you want it is another
question: a set of rows loses their order and silently merges any duplicate
rows, which is either the point or a bug.

```python
print({p for p in pairs})

# This is SUPPOSED to raise. A set stores its items by hash, exactly as a
# dictionary stores its keys -- so a set cannot hold lists either. Tuples
# are fine; lists are not. {tuple(row) for row in matrix} would work.
print({row for row in matrix})
```
