PART A — Building sets

### Question 1

Warm-up — building sets. -> `letters` holds `i`, `m`, `p`, `s` with length `4`; `evens` is `{2, 4, 6, 8, 10}`; `seen` has length `1` after both adds.

Eleven characters went into `letters` and four came out. `set("word")`
iterates the string character by character — the same behaviour that made
`extend("salt")` a trap in worksheet 01 Q6, here being exactly what you
wanted.

The second `add("x")` changed nothing and raised nothing. Adding something
already present is a no-op, which is why you never have to check first.

```python
letters = set("mississippi")
print(letters, len(letters))   # 11 characters in, 4 distinct out

evens = set(range(2, 11, 2))
print(evens)

seen = set()
seen.add("x")
print(len(seen))
seen.add("x")      # already there -- no error, no change
print(len(seen))
```

PART B — Comparing two collections

### Question 3

Comparing two days. -> union has 6 names; intersection is `{bo, dee}`; Monday-only is `{ada, cai}`; Tuesday-only is `{eve, fay}`; symmetric difference is all four of those.

Six distinct people across two days of four visitors each, so exactly two
people came twice.

Note Monday-only and Tuesday-only are different sets, and together they
make up the symmetric difference: `a ^ b` is precisely `(a - b) | (b - a)`.
That is worth knowing so you can reach for one operation instead of two.

```python
print(monday_visitors | tuesday_visitors)   # anyone, either day
print(monday_visitors & tuesday_visitors)   # both days
print(monday_visitors - tuesday_visitors)   # Monday only
print(tuesday_visitors - monday_visitors)   # Tuesday only
print(monday_visitors ^ tuesday_visitors)   # exactly one of the two
```

### Question 4

Hiring questions as set operations. -> `True`, `{'docker'}`, `True`.

"Has every required skill" is `issubset` — answered in one call, with no
loop, no counter and no early-exit logic. "What extras do they bring" is a
difference. "Are these two skill sets unrelated" is `isdisjoint`.

Phrasing a requirement as a set operation is usually shorter AND less wrong
than looping, because the awkward cases — empty sets, duplicates, order —
are handled by definition rather than by you remembering them.

```python
print(required_skills.issubset(applicant_skills))     # all required covered?
print(applicant_skills - required_skills)             # extras they bring
print(required_skills.isdisjoint({"java", "scala"}))  # nothing in common?
```

### Question 5

Two ways to ask one question. -> `True True True`.

`a.issuperset(b)` and `b.issubset(a)` are one relationship viewed from
either end, so they can never disagree. Pick whichever makes the sentence
read correctly where you are standing.

```python
a = applicant_skills.issuperset(required_skills)
b = required_skills.issubset(applicant_skills)
print(a, b, a == b)

# Two ways of asking one question: "does mine contain all of yours" and
# "is all of mine inside yours" must always give the same answer.
```

PART C — Changing sets, and the order problem

### Question 6

Changing a set. -> grows to 5 after `add`, then 7 after `update`, unchanged by `discard`, then 6 after `remove("ada")`.

`update` was handed a LIST and worked fine — it accepts any iterable,
unlike the `|` operator, which insists both sides are already sets.

`discard("nobody")` did nothing and said nothing. That silence is the
entire difference from `remove`, which Q7 demonstrates on the next line.

```python
monday_visitors.add("gus")
print(monday_visitors)

monday_visitors.update(["hal", "ivy"])   # a list is fine
print(monday_visitors)

monday_visitors.discard("nobody")        # absent, and discard shrugs
print(monday_visitors)

monday_visitors.remove("ada")            # present, so this works
print(monday_visitors)
```

The choice is about intent, not convenience. Use `remove` when the item's

Reaching for `discard` to quieten an error you have not understood is how a
real bug gets buried.

### Question 8

Deduplicating, with and without order. -> `6`, then 3 distinct tags, then `['data', 'python', 'sql']` in first-seen order.

`set(tags)` gives you the count instantly and scrambles the sequence. The
loop keeps first-seen order by remembering what it has already emitted.

And `seen` is a SET on purpose: the only question ever asked of it is "have
I seen this before", which is the one thing sets are fastest at. Using a
list there would still be correct and would get slower as the data grew.

This loop is the standard answer to "deduplicate but keep the order", and
it is worth being able to write from memory.

```python
print(len(tags))

distinct = set(tags)
print(distinct, len(distinct))

ordered_unique = []
seen = set()
for tag in tags:
    if tag not in seen:      # membership test is what a set is FOR
        ordered_unique.append(tag)
        seen.add(tag)
print(ordered_unique)
```

### Question 9

Stretch — set comprehensions. -> `{3, 4, 6}`, then `{'data', 'python'}`, then 8 distinct initials.

The first returned 3 values from 6 tags, because several tags share a
length and the set collapsed them.

The second is doing two reductions at once: the `if` dropped `sql` for
being three characters, and the set collapsed the repeated `data` and
`python`. Only one of those two is visible in the code, which is worth
keeping in mind whenever a comprehension returns fewer items than you
expected.

```python
print({len(tag) for tag in tags})

print({tag for tag in tags if len(tag) > 3})

print({name[0] for name in monday_visitors | tuesday_visitors})
```
