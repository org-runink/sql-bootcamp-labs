### Create a list
- named `numbers` containing integers from 1 to 5.
- named `fruits` containing strings `"apple", "banana", and "orange"`.
- named `mixed_list` containing a mix of integers, strings, and boolean values.
- nested list named `nested_list` containing three sublists.

```python
numbers = [1, 2, 3, 4, 5]
fruits = ["apple", "banana", "orange"]
mixed_list = [1, "apple", True, 3.14]
nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
```

### Access and print
- the first element of the `numbers` list.
- the second element of the `fruits` list.
- the last element of the `mixed_list` list.

```python
print(numbers[0])
print(fruits[1])
print(mixed_list[-1])
```

### Access and print the third element of the second sublist in `nested_list`.

```python
print(nested_list[1][2])
```

### Slice the
- `numbers` list to get elements at indices 1, 2, and 3, and print them.
- `fruits` list to get the first two elements and the last two elements, and print them.

```python
print(numbers[1:4])
print(fruits[:2])
print(fruits[-2:])
```

### Update the first element of the `fruits` list to `"grape"`.

```python
fruits[0] = "grape"
print(fruits)
```

### Append `"kiwi"` to the `fruits` list.

```python
fruits.append("kiwi")
print(fruits)
```

### Insert `"pineapple"` at index 1 of the `fruits` list.

```python
fruits.insert(1, "pineapple")
print(fruits)
```

### Extend the `fruits` list with `["blueberry", "watermelon"]`.

```python
fruits.extend(["blueberry", "watermelon"])
print(fruits)
```

### Remove `"banana"` from the `fruits` list.

```python
fruits.remove("banana")
print(fruits)
```

### Pop the last element from the `fruits` list and print it.

```python
popped = fruits.pop()
print(popped)
print(fruits)
```

### Delete the first element of the `fruits` list.

```python
del fruits[0]
print(fruits)
```

### Concatenate the `fruits` list with `["orange", "apple"]` and assign the result to `all_fruits`.

```python
all_fruits = fruits + ["orange", "apple"]
print(all_fruits)
```

### Repeat the `fruits` list twice and print the result.

```python
print(fruits * 2)
```

### Get the length of the `fruits` list and print it.

```python
print(len(fruits))
```

### Sort the `numbers` list in ascending order and print the result.

```python
numbers.sort()
print(numbers)
```

### Reverse the `numbers` list and print the result.

```python
numbers.reverse()
print(numbers)
```

### Find the index of `"apple"` in the `all_fruits` list and print it.

```python
print(all_fruits.index("apple"))
```

### Count the occurrences of `"apple"` in the `all_fruits` list and print it.

```python
print(all_fruits.count("apple"))
```

### Check whether `"kiwi"` and `"mango"` are each in the `fruits` list, and print both results.

```python
print("kiwi" in fruits)
print("mango" in fruits)
```

### Demonstrate the difference between aliasing and copying a list:
- assign `fruits` to a new variable, append an item to it, and show that `fruits` changed too.
- make an actual copy of `fruits` using `.copy()`, append a different item to the copy, and show that the original `fruits` did not change.

```python
# Aliasing: alias points to the SAME list as fruits
alias = fruits
alias.append("mango")
print(fruits)  # changed too, alias and fruits are the same list

# Copying: fruits_copy is an INDEPENDENT list
fruits_copy = fruits.copy()
fruits_copy.append("dragonfruit")
print(fruits)       # unaffected by the change to fruits_copy
print(fruits_copy)
```

### Use a list comprehension to create a list named `fruit_lengths` containing the length of each string in the `fruits` list, and print it.

```python
fruit_lengths = [len(fruit) for fruit in fruits]
print(fruit_lengths)
```
