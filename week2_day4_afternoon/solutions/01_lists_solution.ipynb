### Question 2

Indexing, forwards and backwards. -> `milk`, `bread`, `coffee`, `apples`.

The FOURTH item is at index **3**, because counting starts at 0. This
off-by-one is the most common beginner error in the whole lecture.
`[-1]` is the last item and `[-2]` the one before it, with no need to know
how long the list is — which is why negative indexing is worth the habit.

```python
print(grocery_list[0])    # first item
print(grocery_list[3])    # FOURTH item lives at index 3
print(grocery_list[-1])   # last
print(grocery_list[-2])   # second to last
```

### Question 3

Slicing. -> `['milk', 'beef']`, `['bread', 'apples']`, `['bread', 'apples', 'coffee']`.

`[3:5]` returns TWO items, not three: the end index is excluded. Omitting
the start means "from the beginning" and omitting the end means "to the
end". Every slice returns a NEW list, so slicing never disturbs the
original.

```python
print(grocery_list[:2])    # start omitted = from the beginning
print(grocery_list[3:5])   # 5 is excluded
print(grocery_list[-3:])   # end omitted = to the end
```

### Question 4

Nested lists. -> `['salt', 'pepper', 'chilli']`, then `chilli`, then `['chilli']`.

Compare the last two. `condiments[2][2]` INDEXES the inner list, so you get
the element itself — a plain string. `condiments[2][2:]` SLICES it, so you
get a list that happens to contain one element. Index gives you the thing;
slice gives you a list holding the thing.

```python
print(condiments[2])       # the inner list itself
print(condiments[2][2])    # index into the inner list
print(condiments[2][2:])   # a slice of the inner list
```

### Question 5

Item assignment. -> `['oat milk', 'beef', 'eggs', 'bread', 'apples', 'tea']`.

Nothing was copied or rebuilt: the same list object changed. That is what
"mutable" means, and it is precisely what a tuple will refuse to do in

```python
grocery_list[0] = "oat milk"
grocery_list[-1] = "tea"
print(grocery_list)
```

### Question 6

Growing a list. -> after `append`, `[..., 'tea', 'rice']` (7 items); after `insert(1, 'butter')`, `['oat milk', 'butter', 'beef', ...]` (8); after `extend`, `['oat milk', 'butter', 'beef', 'eggs', 'bread', 'apples', 'tea', 'rice', 'salt', 'pepper']` (10). The trap prints `['milk', 's', 'a', 'l', 't']`.

APPEND VS EXTEND, THE WHOLE DIFFERENCE: `append` adds its argument as ONE
item, whatever that argument is — `append(['salt', 'pepper'])` would have
nested a list inside the list, giving 9 items with a list as the last one.
`extend` iterates its argument and adds each element separately. Iterating
a string yields its characters, which is why `extend('salt')` added four
items. Nothing here is an error, which is what makes it dangerous: you get
a plausible-looking list that is silently wrong.

```python
grocery_list.append("rice")            # one item on the end
print(grocery_list)

grocery_list.insert(1, "butter")       # one item at a chosen position
print(grocery_list)

grocery_list.extend(["salt", "pepper"])  # every item of another list
print(grocery_list)

# extend() iterates its argument, and iterating a string yields characters
trap = ["milk"]
trap.extend("salt")
print(trap)
```

### Question 7

Shrinking a list. -> after `remove('eggs')`, 9 items; `pop(0)` prints `popped: oat milk` and leaves 8; `del` leaves `['beef', 'bread', 'apples', 'tea', 'rice', 'salt', 'pepper']` (7). `dupes` becomes `['beef', 'milk']`.

THREE WAYS TO REMOVE, AND WHEN TO USE EACH: `remove(value)` deletes by
value, and raises `ValueError` if the value is not there. `pop(index)`
deletes by position and RETURNS the item — the only one of the three that
hands the value back, so it is what you want when you still need it.
`del list[index]` deletes by position and returns nothing.

And note `dupes` still contains a `milk`. `remove` stopped at the first
match; it is not a "delete all" operation.

```python
grocery_list.remove("eggs")     # removes by VALUE
print(grocery_list)

popped = grocery_list.pop(0)    # pop RETURNS the item it removed
print("popped:", popped)
print(grocery_list)

del grocery_list[0]             # del removes by INDEX and returns nothing
print(grocery_list)

# remove() stops at the first match
dupes = ["milk", "beef", "milk"]
dupes.remove("milk")
print(dupes)
```

PART C — Operators, ordering, and two traps

### Question 8

Operators and search methods. -> `7`; `pantry` is `['beef', 'bread', 'apples', 'tea', 'rice', 'salt', 'pepper', 'rice', 'salt']`; `['milk', 'milk', 'milk']`; `True`; `True`; `5`; `2`.

`+` and `*` both build a NEW list and leave `grocery_list` untouched —
unlike `extend` in Q6, which changed it. If you find yourself writing
`my_list = my_list + [x]` in a loop, `append` is what you actually want.

`index` reports the first `salt` at position 5 even though `count` finds
two of them: the same first-match-only rule as `remove` in Q7.

```python
print(len(grocery_list))

pantry = grocery_list + ["rice", "salt"]   # + builds a NEW list
print(pantry)

print(["milk"] * 3)                        # * repeats the contents

print("beef" in grocery_list)
print("caviar" not in grocery_list)

print(pantry.index("salt"))   # position of the FIRST match
print(pantry.count("salt"))   # how many matches there are
```

### Question 9

Sorting in place. -> `[1, 3, 5, 7, 9]`, then `[9, 7, 5, 3, 1]`, then `None`, then `[1, 3, 5, 7, 9]`.

THE `None` TRAP: the third print shows `None` because `sort()` returns
nothing — yet the fourth print shows the list really is sorted again. Both
are true at once: the call did the work in place AND handed back `None`.

So `numbers = numbers.sort()` would sort the list and then immediately
overwrite the name with `None`, throwing the data away. Either sort in
place and keep using the same name, or use `sorted(numbers)` — worksheet 05
— when you want a new list and the original left alone.

```python
numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)

# sort() mutates in place and returns None -- there is nothing to assign
print(numbers.sort())
print(numbers)
# so `numbers = numbers.sort()` would throw the list away and leave
# numbers set to None. Sort in place, or use sorted() for a new list.
```

### Question 10

Aliasing vs a real copy. -> `original` and `alias` are BOTH `[1, 2, 3, 4]`. After the copy, `original` is still `[1, 2, 3, 4]` while `true_copy` is `[1, 2, 3, 4, 5]`.

`alias = original` copies the NAME, not the list. There is one list with
two names, so appending through either name shows up under both — you never
touched `original`, and it changed anyway. `copy()` builds a genuine second
list, which is why `true_copy` could grow without disturbing the first.
`original[:]` and `list(original)` do the same job.

This is the single most common source of "my data changed and I never wrote
to it" in Python, and it applies to every mutable structure — including the
dictionaries in worksheet 04.

```python
original = [1, 2, 3]
alias = original       # NOT a copy -- a second name for the SAME list
alias.append(4)
print("original:", original)
print("alias:   ", alias)

true_copy = original.copy()   # original[:] and list(original) do the same
true_copy.append(5)
print("original: ", original)
print("true_copy:", true_copy)
# Appending to alias changed original too, because there is only one list.
# Appending to true_copy did not, because copy() built a second list.
```

### Question 11

Stretch — comprehensions. -> `[4, 5, 6, 3, 4, 4, 6, 4, 4]`, `['bread', 'apples', 'pepper']`, `['BEEF', 'BREAD', 'APPLES', 'TEA', 'RICE', 'SALT', 'PEPPER', 'RICE', 'SALT']`.

Read one left to right: `[what to keep  for each item in the source  if it
qualifies]`. Without an `if`, a comprehension returns exactly one output per
input — which is why `lengths` and `shouted` have 9 entries, the same as
`pantry`. With an `if`, it returns fewer: `long_items` kept 3 of the 9.

All three replace a three-line "make an empty list, loop, append" and read
better for it. They are still just loops, so do not reach for one when the
expression stops fitting on a line.

```python
lengths = [len(item) for item in pantry]
print(lengths)

long_items = [item for item in pantry if len(item) > 4]
print(long_items)

shouted = [item.upper() for item in pantry]
print(shouted)
```
