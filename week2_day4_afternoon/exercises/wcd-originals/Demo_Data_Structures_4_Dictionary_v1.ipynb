# Part 04 - Dictionaries

## Creating Dictionaries

````python
# Creating dictionaries
person = {"name": "John", "age": 30, "city": "New York"}

# Using dict() constructor
car = dict(make="Toyota", model="Camry", year=2018)

keys = ["name", "age", "city"]
my_dict = dict.fromkeys(keys, "Unknown")
print(my_dict)
````

**What you should see:** `{'name': 'Unknown', 'age': 'Unknown', 'city': 'Unknown'}`. `dict.fromkeys(keys, value)` builds a new dictionary using each item in `keys` as a key, all mapped to the same starting `value`, a quick way to initialize a dictionary before filling in real values later.

## Accessing Elements

````python
# Accessing elements
person_name = person["name"]
car_make = car.get("make")  # Using get() method to avoid KeyError

print("Person Name:", person_name)
print("Car Make:", car_make)
````

**What you should see:** `Person Name: John` and `Car Make: Toyota`. Square brackets (`person["name"]`) and `.get()` both retrieve a value by key, the difference shows up when the key is missing, see the next cell.

## Accessing a Missing Key

````python
# Square-bracket access raises an error if the key doesn't exist
person["email"]
````

**What you should see:** a `KeyError: 'email'`. This is exactly why the previous cell used `.get()` for `car_make`, `person.get("email")` would instead return `None` (or a default value you supply, e.g. `person.get("email", "not provided")`) without raising an error.

## Iterating Over a Dictionary

````python
# Loop over just the keys
for key in person:
    print(key)

print()

# Loop over key-value pairs together
for key, value in person.items():
    print(f"{key}: {value}")
````

**What you should see:**
```
name
age
city

name: John
age: 30
city: New York
```
Looping over a dictionary directly (`for key in person`) yields its keys one at a time, in the order they were inserted. `.items()` yields `(key, value)` pairs together, which you can unpack directly in the `for` statement, the most common way to process every entry in a dictionary.

## Modifying Dictionaries

````python
# Modifying dictionaries
person["occupation"] = "Engineer"  # Adding a new key-value pair
car["year"] = 2020  # Updating the value of an existing key
del car["model"]  # Removing a key-value pair

print("Updated Person:", person)
print("Updated Car:", car)
````

**What you should see:** `Updated Person: {'name': 'John', 'age': 30, 'city': 'New York', 'occupation': 'Engineer'}` and `Updated Car: {'make': 'Toyota', 'year': 2020}`. Assigning to a new key adds it, assigning to an existing key overwrites its value, and `del` removes a key-value pair entirely.

## Dictionary Operations

````python
# Dictionary operations
has_age_key = "age" in person  # Check if a key exists
all_keys = person.keys()  # Get all keys
all_values = person.values()  # Get all values
num_entries = len(person)  # Get the number of key-value pairs

print("Has Age Key:", has_age_key)
print("All Keys:", all_keys)
print("All Values:", all_values)
print("Number of Entries:", num_entries)
````

**What you should see:** `Has Age Key: True`, `All Keys: dict_keys(['name', 'age', 'city', 'occupation'])`, `All Values: dict_values(['John', 30, 'New York', 'Engineer'])`, `Number of Entries: 4`. `in` checks the dictionary's *keys*, not its values. `.keys()` and `.values()` return special view objects, not plain lists, they stay in sync with the dictionary if it changes later, but you can wrap either in `list(...)` if you need an actual list.

## Other Methods

````python
# Dictionary methods
car_copy = car.copy()  # Create a shallow copy of the dictionary
car.clear()  # Remove all key-value pairs from the dictionary
year = car_copy.pop("year")  # Remove and return the value of the specified key

print("Car:", car)
print("Car Copy:", car_copy)
print("Year:", year)
````

**What you should see:** `Car: {}`, `Car Copy: {'make': 'Toyota', 'year': 2020}`, `Year: 2020`. `.copy()` makes an independent shallow copy, so `car.clear()` empties only the original `car`, `car_copy` is unaffected. `.pop(key)` removes that key from `car_copy` and returns its value, similar to a list's `.pop()`, but by key instead of by position.

## The `update()` Method

````python
my_dict = {"name": "John", "age": 30}
my_dict.update({"city": "New York", "occupation": "Engineer"})
print(my_dict)
````

**What you should see:** `{'name': 'John', 'age': 30, 'city': 'New York', 'occupation': 'Engineer'}`. `.update()` merges another dictionary (or any iterable of key-value pairs) into this one in place: new keys are added, and any keys that already exist have their values overwritten.

## Dictionary Comprehensions

````python
# A concise way to build a new dictionary by transforming or filtering another iterable
squares = {n: n ** 2 for n in range(1, 6)}
long_words = {word: len(word) for word in ['data', 'engineering', 'sql'] if len(word) > 3}

print("Squares:", squares)
print("Long Words:", long_words)
````

**What you should see:** `Squares: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}` and `Long Words: {'data': 4, 'engineering': 11}` (`'sql'` is excluded since its length, `3`, isn't greater than `3`). `{key_expr: value_expr for item in iterable if condition}` builds a dictionary the same way a list or set comprehension builds those structures.
