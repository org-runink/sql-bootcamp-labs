# Part 05 - Some Builtin Functions

## Discovering What's Available: `dir()`

````python
dir(__builtins__)
````

**What you should see:** a long alphabetical list of names, these are all the built-in functions, types, and exceptions available in Python without needing to `import` anything (`print`, `len`, `int`, `ValueError`, and so on). `dir()` is a handy way to discover what's available on any object or module, not just `__builtins__`.

## `print()`

````python
print("Hello, World!")
````

**What you should see:** `Hello, World!`. `print()` writes text to the output, it's the most commonly used built-in function for seeing what your code is doing.

## `len()`

````python
my_list = [1, 2, 3, 4, 5]
print(len(my_list))
````

**What you should see:** `5`. `len()` returns the number of items in a list, string, tuple, dictionary, set, or any other sized collection.

## `type()`

````python
x = 5
print(type(x))
````

**What you should see:** `<class 'int'>`. `type()` returns an object's type, useful for checking what kind of value you're working with, especially while debugging.

## Type Conversion: `int()`, `float()`, `str()`

````python
num_str = "123"
num_int = int(num_str)      # string -> int
num_float = float(num_str)  # string -> float
str_num = str(num_int)      # int -> string

print(num_int, type(num_int))
print(num_float, type(num_float))
print(str_num, type(str_num))
````

**What you should see:**
```
123 <class 'int'>
123.0 <class 'float'>
123 <class 'str'>
```
These functions convert a value from one type to another where the conversion makes sense: a numeric-looking string can become an `int` or `float`, and any value can be turned into its `str` representation. `int("abc")` would instead raise a `ValueError`, since `"abc"` isn't a valid number.

## `max()`, `min()`, `sum()`

````python
numbers = [3, 1, 4, 2, 5]
print(max(numbers))
print(min(numbers))
print(sum(numbers))
````

**What you should see:** `5`, `1`, `15`. `max()` and `min()` find the largest and smallest item in an iterable, `sum()` adds them all up. All three work on any iterable of comparable (or, for `sum`, addable) values, not just lists.

## `sorted()`

````python
numbers = [3, 1, 4, 2, 5]
sorted_numbers = sorted(numbers)
print(sorted_numbers)
````

**What you should see:** `[1, 2, 3, 4, 5]`. Unlike a list's own `.sort()` method, `sorted()` returns a *new* sorted list and leaves the original `numbers` unchanged.

````python
# sorted() also accepts reverse= and a key= function for custom sort order
print(sorted(numbers, reverse=True))

words = ['banana', 'kiwi', 'apple', 'fig']
print(sorted(words, key=len))  # sort by string length instead of alphabetically
````

**What you should see:** `[5, 4, 3, 2, 1]` and `['kiwi', 'fig', 'banana', 'apple']`. `reverse=True` sorts largest-to-smallest, and `key=` lets you sort by something other than the values themselves, here, by each word's length instead of alphabetically.

## Lazy Iterators: `reversed()`, `range()`, `zip()`, `enumerate()`

These functions don't return a list, they return generator-like objects or iterators, which produce values lazily, as needed. Printing one directly shows its type rather than its contents; wrap it in `list()` (or loop over it) to see the values.

````python
# return generator-like objects or iterators, which produce values lazily, as needed.
reversed(numbers)
````

````python
list(reversed(numbers))
````

**What you should see:** first a `<list_reverseiterator ...>` object (the exact memory address will differ every run), then `[5, 2, 4, 1, 3]` once wrapped in `list()`, `numbers` in reverse order.

````python
# return generator-like objects or iterators, which produce values lazily, as needed.
range(0, 5)
````

````python
list(range(0, 5))
````

**What you should see:** first `range(0, 5)` (a `range` object remembers its start/stop/step rather than storing every number), then `[0, 1, 2, 3, 4]` once wrapped in `list()`. Note that `5` itself is excluded, `range(0, 5)` stops before reaching `5`, just like list slicing.

````python
# return generator-like objects or iterators, which produce values lazily, as needed.
names = ['John', 'Alice', 'Bob']
ages = [30, 25, 35]
zip(names, ages)
````

````python
list(zip(names, ages))
````

**What you should see:** first a `<zip ...>` object, then `[('John', 30), ('Alice', 25), ('Bob', 35)]` once wrapped in `list()`. `zip()` pairs up corresponding elements from two (or more) iterables, stopping as soon as the shortest one runs out.

````python
# return generator-like objects or iterators, which produce values lazily, as needed.
enumerate(names)
````

````python
list(enumerate(names))
````

**What you should see:** first an `<enumerate ...>` object, then `[(0, 'John'), (1, 'Alice'), (2, 'Bob')]` once wrapped in `list()`. `enumerate()` pairs each item with its index, most often used in a `for index, value in enumerate(items):` loop when you need both the position and the value.

## `map()` and `filter()`

Like `reversed()`, `range()`, `zip()`, and `enumerate()`, both of these also return lazy iterator objects, wrap them in `list()` to see their contents.

````python
numbers = [1, 2, 3, 4, 5]

# map() applies a function to every item in an iterable
squared = list(map(lambda n: n ** 2, numbers))

# filter() keeps only the items for which a function returns True
evens = list(filter(lambda n: n % 2 == 0, numbers))

print("Squared:", squared)
print("Evens:", evens)
````

**What you should see:** `Squared: [1, 4, 9, 16, 25]` and `Evens: [2, 4]`. `map(function, iterable)` returns the result of calling `function` on every item, `filter(function, iterable)` keeps only the items where `function` returns a truthy value. Both are commonly written with a `lambda` (a small, unnamed function) as shown here, though a list comprehension (`[n ** 2 for n in numbers]`) often reads more clearly for simple cases like this one.
