# Part 02 - Tuples

## Creating Tuples

````python
# Creating tuples
coordinates = (3, 4)
colors = ("red", "green", "blue")
mixed_tuple = (1, "apple", True, 3.14)

# Single-element tuple
single_element_tuple = (5,)  # Note the comma after the single element

# Nested tuple
nested_tuple = ((1, 2), (3, 4), (5, 6))
````

**What you should see:** no output, this cell only creates variables. Note the trailing comma in `single_element_tuple = (5,)`, without it, `(5)` would just be the integer `5` in parentheses, not a tuple. `nested_tuple` shows that a tuple can contain other tuples, just like a list can contain other lists.

## Accessing Elements

````python
# Accessing elements
x_coordinate = coordinates[0]
second_color = colors[1]
last_element = mixed_tuple[-1]

print("X Coordinate:", x_coordinate)
print("Second Color:", second_color)
print("Last Element:", last_element)
````

**What you should see:**
```
X Coordinate: 3
Second Color: green
Last Element: 3.14
```
Tuples support the same indexing as lists, `coordinates[0]` is the first item, and `mixed_tuple[-1]` is the last item counting from the end.

## Tuples Are Immutable

````python
# Attempt to change an element of a tuple
coordinates[0] = 10
````

**What you should see:** a `TypeError: 'tuple' object does not support item assignment`. This is the key difference from a list: once created, a tuple's elements can't be changed, added, or removed. That immutability makes tuples useful for data that shouldn't change (like fixed coordinates) and means they can be used as dictionary keys or set elements, which lists cannot.

## Slicing

````python
# Slicing
some_coordinates = coordinates[1:]  # Get elements starting from index 1
first_two_colors = colors[:2]  # Get elements from start to index 1

print("Some Coordinates:", some_coordinates)
print("First Two Colors:", first_two_colors)
````

**What you should see:**
```
Some Coordinates: (4,)
First Two Colors: ('red', 'green')
```
Slicing a tuple works exactly like slicing a list, `coordinates[1:]` returns everything from index 1 onward, but the result is itself a tuple, not a list.

## Tuple Unpacking

````python
# Unpack a tuple's elements directly into named variables
x, y = coordinates
print("x:", x)
print("y:", y)

color1, color2, color3 = colors
print(color1, color2, color3)

# Use * to collect "the rest" of the values into a list
first_color, *remaining_colors = colors
print("first_color:", first_color)
print("remaining_colors:", remaining_colors)
````

**What you should see:**
```
x: 3
y: 4
red green blue
first_color: red
remaining_colors: ['green', 'blue']
```
Unpacking assigns each element of the tuple to a variable in one line, the number of variables on the left must match the number of elements (unless you use `*` to collect any extras into a list, as with `*remaining_colors` above).

## Working with Nested Tuples

````python
# A tuple of tuples: use two sets of square brackets to reach an inner element
print(nested_tuple)
print(nested_tuple[1])       # the second inner tuple
print(nested_tuple[1][0])    # the first element of the second inner tuple
````

**What you should see:**
```
((1, 2), (3, 4), (5, 6))
(3, 4)
3
```
Just like nested lists, `nested_tuple[1]` selects the second inner tuple, and chaining a second index reaches into that inner tuple.

## Tuple Operations

````python
# Tuple operations
all_colors = colors + ("yellow", "purple")  # Concatenation
repeated_tuple = colors * 2  # Repetition
num_colors = len(colors)  # Length of the tuple

print("All Colors:", all_colors)
print("Repeated Tuple:", repeated_tuple)
print("Number of Colors:", num_colors)
````

**What you should see:**
```
All Colors: ('red', 'green', 'blue', 'yellow', 'purple')
Repeated Tuple: ('red', 'green', 'blue', 'red', 'green', 'blue')
Number of Colors: 3
```
`+` concatenates two tuples into a new tuple, `*` repeats a tuple's contents, and `len()` returns the number of elements, all without modifying the original `colors` tuple, since tuples can't be modified in place.

## Tuple Methods

````python
# Tuple methods
index_of_red = colors.index("red")  # Find the index of an element
count_red = colors.count("red")  # Count occurrences of an element

print("Index of Red:", index_of_red)
print("Count of Red:", count_red)
````

**What you should see:** `Index of Red: 0`, `Count of Red: 1`. Tuples only have two built-in methods, `index()` and `count()`, unlike lists, they have no `append()`, `remove()`, `sort()`, etc., because those would all modify the tuple in place.
