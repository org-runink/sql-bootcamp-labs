# Week 4, day 3 (morning) — Worksheet 01 SOLUTIONS: What goes wrong without governance

This is the lecture's opening story, as a worksheet. Five departments, one
popular dataset, and no governance.

> *Discussion, not coding. There is no dataset and no SQL here.*

**The story.** Department **B** owns Dataset B: it collects and cleans the data,
uses it for reporting, and is responsible for infrastructure security, privacy
compliance and data quality. B gives **A** access — still manageable. The data
becomes popular and more departments want it. **C** gets a copy. **E** asks
**C** — not B — for a copy of its own.

Everyone seems happy. Real life is not.

**These are discussion answers, not a marking scheme.** Where a question asks for a judgement, the reasoning matters more than matching what is written here — a group that disagrees well has done the exercise.

STEP 1 — predict the failures

### Question 1

Before reading the lecture's answer: **what could go wrong** when data is
copied and shared across many teams without clear governance? List as many
distinct failures as you can in five minutes.
> **NOTE:** aim for failures that are *different in kind*, not five versions of
> "someone sees data they should not".

The lecture names three, and a group usually finds most of them:

**1. The data is not properly protected.** B encrypted it; the copies may not
be. Every hop is a new place the data can leak from, under a team that never
agreed to B's security obligations.

**2. The data is not up to date.** B changes the data and tells nobody. C is
still serving an old copy; D was never notified. Two departments now report
different numbers and both believe they are right.

**3. The data is not reliable.** E's copy came from C, not B. Whatever C
filtered, joined or fixed is now baked in — E has inherited C's assumptions
without knowing they exist, and the data may be biased.

Groups often add: nobody knows who to ask when a number looks wrong; the same
dataset exists five times at five different costs; and B is still formally
accountable for data it no longer controls.

STEP 2 — trace the accountability

### Question 2

Follow one hop: **E asked C, not B.** Who is accountable for the data E is
using — for its accuracy, its privacy compliance, and its security? Name the
party for each, and say how you know.

The uncomfortable answer is that **accountability did not move, but control
did**.

B is still the owner: it collects, cleans and is responsible for security,
privacy compliance and quality. But B does not know E exists. B cannot notify E
of a change, cannot audit E's storage, and cannot revoke E's access — because B
never granted it.

C did not take on ownership either. C made a copy for its own use and passed it
along; it has no process for maintaining it, and probably no mandate to.

So the honest answer is *nobody*, which is the point of the story. This is the
gap governance closes: an owner who is known, access that is granted rather than
copied, and a change that reaches everyone using the data.

STEP 3 — from story to definition

### Question 3

Based on the story, **what do you think data governance means?** Write two
or three sentences in your own words. Do not look up a definition first.

A workable answer names both halves — the *what* and the *who*.

Data governance is the set of decisions and accountabilities that make data
trustworthy and safe to use: who owns each dataset, what it means, who may
access it, how quality is checked, and how changes are communicated.

The distinction worth landing: governance is mostly about **decision rights and
accountability**, not tools. The story's failures were not caused by missing
technology — B had encryption. They were caused by nobody being responsible for
the copies. Tools enforce governance; they do not constitute it.

STEP 4 — the fix, without buying anything

### Question 4

Suppose you cannot buy any new tooling. Name **three changes** to how these
departments work that would prevent most of the failures above, and say which
failure each one addresses.

Answers vary, but the strong ones are process changes, not technology:

**Grant access to the source; forbid copies.** If everyone reads B's dataset,
the stale-copy and inherited-bias failures largely disappear, and B regains the
ability to revoke.

**Name an owner and publish it.** Every dataset has one accountable team, and
consumers know who to ask. This fixes "nobody knows who to ask", and makes
change notification somebody's job.

**Register consumers.** B keeps a list of who uses the data. Without it, B
cannot notify anyone of a change — which was risk 2 in the story.

Groups often add a shared definition of the fields, which is the next
worksheet's problem.
