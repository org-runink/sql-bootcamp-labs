PART A — The shape of a for loop

### Question 1

Iterating, then unpacking. -> six tuples like `('jack', 'DS101')`, then six lines reading `jack is on DS101`.

A `for` loop hands you one **item** per pass, and here every item is a
2-tuple. `for row in enrollment:` gives you the tuple whole;
`for name, course in enrollment:` splits it into two names as it goes.

Unpacking is not a special loop feature — it is ordinary assignment. The
loop header is doing `name, course = ("jack", "DS101")` on every pass, and
it will raise `ValueError` the moment a row has three elements instead of
two.

```python
for row in enrollment:
    print(row)

print("---")

# Unpacking: the tuple is split into two names as the loop header runs.
for name, course in enrollment:
    print(name, "is on", course)
```

### Question 2

Hand-rolled accumulator. -> `12.16`, then `2`, then `12.16` again from `sum()`.

The accumulator pattern is the loop you will write most often: a variable
set **before** the loop, updated **inside** it, read **after** it. Two of
those three are outside the loop body, and that is entirely down to the
indentation.

Indent `print(total)` by four spaces and it moves inside — you would get
`1.99`, `5.98`, `7.17`, `12.16` and no error at all. Python cannot tell
the difference between a running total you wanted and one you did not.

`sum()` agrees, and is what you would actually write. The hand-rolled
version is worth doing once, because a running *count* with a condition in
it — the `expensive` line — has no one-word built-in.

```python
total = 0
expensive = 0

for price in produce.values():
    total = total + price
    if price > 2.00:
        expensive = expensive + 1

print(total)          # outside the loop -- one line
print(expensive)

print(sum(produce.values()))
```

PART B — range()

### Question 3

The five ranges. -> `[0, 1, 2, 3, 4]`, `[1, 2, 3, 4]`, `[1, 3]`, `[5, 4, 3, 2, 1]`, and **`[]`**.

`range` never reaches its stop value. `range(5)` stops before 5;
`range(1, 5)` stops before 5; `range(5, 0, -1)` stops before 0, which is
why 0 is missing from the countdown.

`range(5, 0)` is empty and raises nothing. With no step given the step is
`+1`, and you cannot walk upwards from 5 and arrive before 0, so the loop
body runs zero times. **An empty range is not an error — it is a loop that
silently does nothing.** If a report comes out blank, this is one of the
first things to check.

```python
print("range(5)       ", list(range(5)))
print("range(1, 5)    ", list(range(1, 5)))
print("range(1, 5, 2) ", list(range(1, 5, 2)))
print("range(5, 0, -1)", list(range(5, 0, -1)))
print("range(5, 0)    ", list(range(5, 0)))   # no step given, so step is +1
```

### Question 4

Indexing by hand, then `enumerate`. -> the same six lines `0 jack` … `5 marie` from both loops. Then `<enumerate object at 0x…>`.

Both loops give identical output, but only one of them can go wrong. The
`range(len(...))` version has to get the length right, index correctly, and
reach back into `enrollment[i]` on every pass; `enumerate` cannot get out
of step with the list because it *is* the list.

Slide 43 does this a third way — a manual `i = 0` with an `if`/`else` where
both branches run `i += 1`. That is `enumerate` written in nine lines
rather than one, and the `if` in it does nothing.

The last line prints an object, not values: `enumerate` is lazy, producing
pairs only as the loop asks for them. The hex address will differ on your
machine; that is the memory location, and it means nothing.

```python
for i in range(len(enrollment)):
    print(i, enrollment[i][0])

print("---")

for i, row in enumerate(enrollment):
    print(i, row[0])

print("---")

# enumerate is lazy: it does not build a list, it hands out pairs on demand.
print(enumerate(enrollment))
```

### Question 5

The slide's "reversal". -> `5 marie`, `3 jen`, `1 lucy` — **3 of 6**. The real reversal prints all **6 of 6**, ending `0 jack`.

Slide 46 is titled *reversing a list* and does not reverse the list. Two
separate faults in one line:

- the step is `-2`, so it skips every other enrolment;
- the stop is `0`, and `range` never reaches its stop, so index 0 —
  `jack` — can never be printed no matter what the step is.

Half the data is missing and nothing raises. The slide even prints its own
three-line output as if that were the whole list.

The correct form is `range(len(x) - 1, -1, -1)`: start at the last index,
step down by one, and stop *below* zero so zero is included. In real code
you would write `for row in reversed(enrollment):` and avoid the indexes
altogether.

```python
num_students = len(enrollment)

# The slide's version, run as written.
count_slide = 0
for i in range(num_students - 1, 0, -2):
    print(i, enrollment[i][0])
    count_slide = count_slide + 1
print("slide version printed", count_slide, "of", num_students)

print("---")

# A real reversal: every index, one at a time, and stop AFTER 0 -- which
# means the stop value has to be -1, because stop is never reached.
count_real = 0
for i in range(num_students - 1, -1, -1):
    print(i, enrollment[i][0])
    count_real = count_real + 1
print("real reversal printed", count_real, "of", num_students)
```

PART C — Looping over a string

### Question 6

Lowercasing a string character by character. -> the 12-character list `['d', 'a', 't', 'a', ' ', …]`, then `data science`.

Looping a string gives you one **character** at a time — the space at index
4 included, since a space is a character like any other.

`"".join(list_of_chars)` is how you get back to a string. The separator is
the string you call it on, so `"".join` glues with nothing and `", ".join`
would put a comma between each character.

All of this is `random_str.lower()` in real code. The point is the shape of
the loop, not the result.

```python
modified_str = []

for char in random_str:
    modified_str.append(char.lower())

print(modified_str)
print("".join(modified_str))
```

### Question 7

Slide 48's index stepping. -> `DaTa SCiEnCe`, and the indexes visited are `[0, 2, 4, 6, 8, 10]` — the **even** ones.

The code and its output are correct and agree with each other. The slide's
**title** says "odd indexes", and `range(0, len, 2)` visits the even ones.
For odd you would start at 1: `range(1, len(char_str), 2)`.

Note the two conversions bracketing the work. Strings are immutable, so
`random_str[0] = "D"` would raise; `list(random_str)` gives you something
you *can* assign into, and `"".join(...)` turns it back. That out-and-back
is the standard move whenever you need to edit a string in place.

```python
char_str = list(random_str)
print(char_str)

for i in range(0, len(char_str), 2):
    char_str[i] = char_str[i].upper()

print("".join(char_str))

# The indexes visited:
print(list(range(0, len(random_str), 2)))
# 0, 2, 4, ... -- these are the EVEN indexes. The slide is titled "odd
# indexes"; its code and its output are both even. range(1, len, 2) would
# be the odd ones.
```

PART D — Looping over a dictionary

### Question 8

Three ways through a dictionary. -> the four product names, then the four prices `1.99 / 3.99 / 1.19 / 4.99`, then four `X costs Y` lines. Then `dict_keys([...])`.

Looping a dictionary directly gives you its **keys** — not its values, and
not pairs. That is the one to remember, because `for x in produce:` looks
like it should hand you the whole entry and does not.

`.items()` with two loop variables is the form you will use most. It is
unpacking again, exactly as in Q1.

`produce.keys()` prints as `dict_keys([...])` because it is a **view**, not
a list: it stays attached to the dictionary and reflects later changes.
Wrap it in `list()` if you need a frozen copy — which is precisely the fix
for Q10.

```python
for k in produce:
    print(k)

print("---")

for v in produce.values():
    print(v)

print("---")

for k, v in produce.items():
    print(k, "costs", v)

print("---")

# A view, not a list -- it tracks the dictionary rather than copying it.
print(produce.keys())
```

Slide 51's 13% price rise. -> `{'Red kidney beans': 2.2487, 'White beans': 4.5087, 'Green lentils': 1.3446999999999998, 'Chopped tomatoes': 5.6387}`

### Question 10

Adding a key mid-loop. -> `RuntimeError: dictionary changed size during iteration`. The `print(len(produce))` never runs.

Python raises on the *second* pass, the moment it notices the dictionary is
a different size from when the loop began. It cannot promise to visit each
key exactly once any more, so rather than quietly skipping or repeating
some, it stops.

This is a rare kindness. Compare Q5 and Q9, where wrong answers came out
silently — here the language refuses to guess.

Two fixes, and they mean different things:
- `for k in list(produce):` iterates a **snapshot** of the keys, so the new
  ones are not visited;
- build a **new** dictionary and leave the original alone. Usually the
  better answer, and a dict comprehension does it in one line (worksheet 03).

```python
for k in produce:
    produce[k + " (organic)"] = 0.0

# Never reached. RuntimeError: dictionary changed size during iteration.
# Python noticed the dictionary grew underneath the loop and refused to
# carry on -- it cannot promise to visit each key exactly once any more.
# Build a NEW dict, or loop over list(produce) to iterate a snapshot.
print(len(produce))
```

### Question 11

Counting occurrences. -> `{'Tuna': 2, 'Salmon': 3, 'Mackerel': 3, 'Pasta': 2, 'Rice': 1, 'Bread': 1}`, then `6 distinct items from 12 entries`.

The count-into-a-dictionary loop, and it is worth memorising: check `in`,
increment if present, start at 1 if not. You will write it constantly.

The `else` branch is doing the real work — a key that does not exist yet
cannot be incremented, and `counts[item] += 1` on a missing key raises
`KeyError`. That is why the `if` is there.

Note what the dictionary gives you for free: `len(counts)` is the number of
**distinct** items, and `len(grocery_list)` is the number of entries. 6
and 12. If those two numbers ever get confused in a report, this is the
distinction that was lost.

```python
counts = {}

for item in grocery_list:
    if item in counts:
        counts[item] = counts[item] + 1
    else:
        counts[item] = 1

print(counts)
print(len(counts), "distinct items from", len(grocery_list), "entries")
```

PART E — Nesting

### Question 12

Flattening a nested list. -> the nine course names in order, then `3 lists -> 9 courses`, then `outer ran 3 times; inner ran 9 times`.

The outer loop hands you a **list**; the inner one hands you a course. If
you only write the outer loop you get three lists, not nine strings — which
is the usual first attempt.

The two counters are the point. The inner body runs 3 × 3 = 9 times, and
that multiplication is the whole cost model of nested loops: 3 lists of 3
is nine passes, 3 lists of 1,000 is three thousand, and a nested loop over
two 10,000-row tables is a hundred million. Loops over a *pair* of
collections is where slow code usually comes from — and in worksheet 07 you
will replace one with a dictionary lookup.

```python
courses_unpacked = []
outer_runs = 0
inner_runs = 0

for course_list in courses:
    outer_runs = outer_runs + 1
    for course in course_list:
        inner_runs = inner_runs + 1
        courses_unpacked.append(course)

print(courses_unpacked)
print(len(courses), "lists ->", len(courses_unpacked), "courses")
print("outer ran", outer_runs, "times; inner ran", inner_runs, "times")
```
