# **Part 02 - Tuples**

---

### Create a
- tuple named `coordinates` containing integers 3 and 4.
- tuple named `colors` containing strings `"red", "green", and "blue"`.
- tuple named `mixed_tuple` containing a mix of integers, strings, and boolean values.
- single-element tuple named `single_element_tuple` containing the integer 5.
- nested tuple named `nested_tuple` containing three tuples.

```python
coordinates = (3, 4)
colors = ("red", "green", "blue")
mixed_tuple = (1, "apple", True, 3.14)
single_element_tuple = (5,)
nested_tuple = ((1, 2), (3, 4), (5, 6))
```

### Access and print the
- first element of the `coordinates` tuple.
- second element of the `colors` tuple.
- last element of the `mixed_tuple` tuple.

```python
print(coordinates[0])
print(colors[1])
print(mixed_tuple[-1])
```

### Attempt to change the first element of the `coordinates` tuple to `10`, and explain why it causes an error.

```python
coordinates[0] = 10
```

This raises a `TypeError: 'tuple' object does not support item assignment`. Tuples are immutable, once created, their elements can't be changed.

### Slice the
- `coordinates` tuple to get elements starting from index 1 and print them.
- `colors` tuple to get the first two elements and print them.

```python
print(coordinates[1:])
print(colors[:2])
```

### Unpack the `coordinates` tuple into two variables, `x` and `y`, and print each one.

```python
x, y = coordinates
print(x)
print(y)
```

### Access and print the first element of the second sublist in `nested_tuple`.

```python
print(nested_tuple[1][0])
```

### Concatenate the `colors` tuple with `("yellow", "purple")` and print the result.

```python
print(colors + ("yellow", "purple"))
```

### Repeat the `colors` tuple twice and print the result.

```python
print(colors * 2)
```

### Get the length of the `colors` tuple and print it.

```python
print(len(colors))
```

### Find the index of `"red"` in the `colors` tuple and print it.

```python
print(colors.index("red"))
```

### Count the occurrences of `"red"` in the `colors` tuple and print it.

```python
print(colors.count("red"))
```
