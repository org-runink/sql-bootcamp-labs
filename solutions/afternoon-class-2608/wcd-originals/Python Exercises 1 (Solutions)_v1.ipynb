### Exercise 8 

- Create a list with the following numbers: `[42, 2, 9, 7, 3, 53, 18]`
- Create a new list with only the even numbers from the first list

```python
# Your code below
first_lst = [42, 2, 9, 7, 3, 53, 18]
first_lst
```

The next three cells solve the same problem three different ways, using a manual loop with `append()`, a manual loop with `remove()`, and a list comprehension, to show how the same result can be reached with increasingly concise code.

```python
# Using the for loop and append
new_lst = [] #Create an empty list
for x in first_lst: #Go through each item in the first_lst
    if x % 2 == 0:
        new_lst.append(x) # add to new_lst for all even numbers
        
new_lst
```

```python
# Using for loop and remove
new_lst = first_lst.copy() # Making a copy of the existing list
for x in first_lst:
    if x % 2 != 0:
        new_lst.remove(x) # Delete the odd numbers
new_lst
```

```python
# Using List Comprehension
[x for x in first_lst if x % 2 == 0]
```

### Exercise 10
Imagine a store has sold 5 orders of its product. Given a list of the prices of each order and a list of quantities for each order, calculate the total sales revenue.

If you'd like a challenge, try solving it in one line. (Hint: list comprehension)

```python
prices = ['$5', '$10', '$1', '$20', '$7']
quantities = [5, 2, 50, 5, 1]

# Your code below
int(prices[0].replace('$',''))*quantities[0]
```

The cell above only computes the revenue for the *first* order (index `0`), it's a quick check that the `$` stripping and conversion works before scaling it up. The next cell loops over every order to get the actual total.

```python
total = 0
for i in range(len(prices)): # i will be the different indexes in the list
    # Convert the prices to integers
    price = int(prices[i].replace('$','')) # Need to remove $ before converting to int
    quantity = quantities[i]
    total += price * quantity

total
```

### Hangman

Create a Hangman game. The rules of Hangman can be found at:  https://en.wikipedia.org/wiki/Hangman_(game). For our version of the game we'll do the following:
- The answer will be set to `'kangaroo'`
- Use the `input` function to get the user's guess
- Display the current board (ka_a_oo) and current guessed letters every turn
- The player has 5 incorrect guess until they lose
- If the user guesses all letters end the program and print 'You win'
- If the user makes 5 incorrect guesses print 'You lose'

### Bonus
Try improving your game. You can make it as complex as you'd like. Some ideas to try:

    - ex. If the user inputs an uppercase letter or a number
- Make the game two players
    - One player creates the answer and the other guesses
    - When they're done switch roles
    - Play multiple rounds and keep score to declare the winner

**Note:** this cell uses `input()` in a loop, so it only works when run interactively in a live Jupyter session. It will pause after each guess waiting for you to type a letter; running it non-interactively (for example, an automated `Run All`) will error on the first `input()` call since there's no one there to respond.

```python
# Your code below
answer = 'kangaroo'

board = ['_'] * len(answer)
guessed_letters = []
max_guesses = 5

# Create a loop that decreases after each wrong guess and ends either with the board equal to the answer or the guesses = 0
while (''.join(board) != answer) and (max_guesses > 0):
    guess = input('Guess a letter: ').lower()
    guessed_letters.append(guess)

    if guess in answer:
        for i in range(len(answer)):
            if guess == answer[i]:
                board[i] = guess
    else:
        max_guesses -= 1

    print('Board:', ''.join(board))            # Show the current game status
    print('Guessed letters:', guessed_letters)  # Show every letter guessed so far
    print('Guesses remaining:', max_guesses)    # Show how many tries are left

if ''.join(board) == answer:
    print('You Win!')
else:
    print('You Lose!')
```
