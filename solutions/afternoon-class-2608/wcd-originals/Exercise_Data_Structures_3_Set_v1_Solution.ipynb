### Create a set named
- `letters` containing characters `'a' through 'e'`.
- `vowels` containing vowels `'a', 'e', 'i', 'o', 'u'`.
- `numbers_set` containing integers from 1 to 5 using `set()` constructor.
- `x`, add integers 1 and 2 to it, and attempt to add 2 again.

```python
letters = {'a', 'b', 'c', 'd', 'e'}
vowels = {'a', 'e', 'i', 'o', 'u'}
numbers_set = set([1, 2, 3, 4, 5])

x = set()
x.add(1)
x.add(2)
x.add(2)  # adding 2 again has no effect, it's already in the set
print(x)
```

### Check if
- `'a'` is a member of the `letters` set and print the result.
- `'f'` is a member of the `letters` set and print the result.

```python
print('a' in letters)
print('f' in letters)
```

### Perform:
- a `union` operation between the `letters` and `vowels` sets and print the result.
- an `intersection` operation between the `letters` and `vowels` sets and print the result.
- a `difference` operation between the `letters` and `vowels` sets and print the result.
- a `symmetric difference` operation between the `letters` and `vowels` sets and print the result.
- `union`, `intersection`, `difference`, and `symmetric difference` operations between the `letters` set and `{'f', 'g'}` using set operators and print the results.

```python
print(letters.union(vowels))
print(letters.intersection(vowels))
print(letters.difference(vowels))
print(letters.symmetric_difference(vowels))

print(letters | {'f', 'g'})
print(letters & {'f', 'g'})
print(letters - {'f', 'g'})
print(letters ^ {'f', 'g'})
```

### Do the following:
- Add `'f'` to the `letters` set,
- update the letters set with `{'g', 'h'}`,
- remove `'a'` from the `letters` set,
- and discard `'z'` if it exists.

```python
letters.add('f')
letters.update({'g', 'h'})
letters.remove('a')
letters.discard('z')
print(letters)
```

### Clear all elements from the `letters` set and print it.

```python
letters.clear()
print(letters)
```

### Create sets `s1`, `s2`, and `s3` containing `{1, 2}`, `{1, 2, 4}`, and `{5}`, respectively.

```python
s1 = {1, 2}
s2 = {1, 2, 4}
s3 = {5}
```

### Check if:
- `s1` and `s2` are disjoint and print the result.
- `s1` and `s3` are disjoint and print the result.
- `s1` is a subset of `s2` and print the result.
- `s2` is a superset of `s1` and print the result.

```python
print(s1.isdisjoint(s2))
print(s1.isdisjoint(s3))
print(s1.issubset(s2))
print(s2.issuperset(s1))
```

### Use a set comprehension to create a set named `letter_lengths` containing the length of each word in the list `['apple', 'banana', 'avocado', 'blueberry']`, and print it.

```python
letter_lengths = {len(word) for word in ['apple', 'banana', 'avocado', 'blueberry']}
print(letter_lengths)
```
