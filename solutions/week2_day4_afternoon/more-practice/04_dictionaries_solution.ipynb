PART A — Building and reading

### Question 1

Warm-up — three ways to build a dictionary. -> `{'tea': 2.5, 'coffee': 3.75}` twice, then `{'a': 0, 'b': 0, 'c': 0}`, then `2`.

`2.50` printed as `2.5` — typed, never stored, the same point as worksheet
01 Q1.

`dict()` with keyword arguments needs every key to be a valid Python name,
so it cannot produce a key with a space in it or one starting with a digit.
The literal form has no such limit, which is why it is the default choice.

```python
prices = {"tea": 2.50, "coffee": 3.75}
prices2 = dict(tea=2.50, coffee=3.75)   # keys become names, so no quotes
counters = dict.fromkeys(["a", "b", "c"], 0)

print(prices)
print(prices2)
print(counters)
print(len(prices))
```

### Question 2

Reading nested config. -> `{'host': 'localhost', 'port': 8888}`, `8888`, `bo`, `True`.

`config["server"]` returns a dict, so the next step is a KEY.
`config["users"]` returns a list, so the next step is an INDEX. The bracket
looks identical in both cases, which is exactly why printing the
intermediate value is the quickest way to unstick yourself in nested data.

```python
print(config["server"])
print(config["server"]["port"])   # dict inside a dict
print(config["users"][1])         # list inside a dict -- index, not key
print(config["debug"])
```

### Question 3

Present-but-zero versus missing. -> `12`, `None`, `0`, `True`, `0`.

READ THE LAST THREE LINES TOGETHER. `stock.get("sprocket", 0)` returned `0`
because the key is absent. `stock["gizmo"]` returned `0` because the key is
present and its value genuinely is zero.

So a `.get()` with a default CANNOT distinguish "missing" from "present and
zero" — and the same applies to empty strings and empty lists. For stock
levels that difference is the whole business question: nothing on the shelf
is not the same as no such product. Use `in` when you need to tell them
apart.

```python
print(stock.get("widget"))
print(stock.get("sprocket"))        # missing -> None
print(stock.get("sprocket", 0))     # missing -> your default

print("gizmo" in stock)   # True: the key exists
print(stock["gizmo"])     # ...and its value happens to be 0

# So `stock.get("gizmo", 0)` and `stock.get("sprocket", 0)` both return 0 for
# completely different reasons. Use `in` when you need to tell them apart.
```

PART B — Changing a dictionary

### Question 5

Changing a dictionary. -> ends at `{'widget': 15, 'gizmo': 3, 'sprocket': 5}`, with `pop` returning `7` and `cog` deleted.

`update` did two different things in one call — replaced `widget` and added
`cog` — with nothing in the output to say which was which.

Note the surviving keys kept their original insertion order: additions go
on the end, and removing `doohickey` did not reshuffle the rest.
Dictionaries have preserved insertion order since Python 3.7, which makes
output predictable but is not a substitute for sorting.

```python
stock["sprocket"] = 5      # new key -> added
print(stock)

stock["gizmo"] = 3         # existing key -> replaced
print(stock)

stock.update({"widget": 15, "cog": 2})   # one update, one add
print(stock)

gone = stock.pop("doohickey")   # returns the value
print("popped:", gone)
print(stock)

del stock["cog"]
print(stock)
```

### Question 6

Views and summaries. -> the three `dict_*` view objects, then `23`, `widget`, `['gizmo', 'sprocket', 'widget']`.

`max(stock, key=stock.get)` reads almost as English: of the keys, the one
whose `stock.get` value is largest. It returns the KEY, not the value — use
`max(stock.values())` if you wanted `15`.

`sorted(stock)` sorts the KEYS, because iterating a dictionary iterates its
keys. It returns a plain list, so the dictionary itself is untouched and
still in insertion order.

```python
print(stock.keys())
print(stock.values())
print(stock.items())

print(sum(stock.values()))
print(max(stock, key=stock.get))   # the KEY whose value is largest
print(sorted(stock))               # sorting a dict sorts its keys
```

PART C — The two jobs you will actually do

Counting and grouping. Both are one loop, an empty dict, and a check for
whether the key exists yet — only the value type changes.

### Question 7

Counting with a dictionary. -> `{'red': 3, 'blue': 2, 'green': 1}`, then `red`.

THE COUNTING PATTERN, and you will write it constantly: empty dict, loop,
create the key at zero if it is missing, add one. Three lines, and it works
for anything that can be a key.

`max(counts, key=counts.get)` finds the most common — but be aware it
returns ONE key even when there is a tie, and says nothing about the tie
existing. Sort the items instead if that distinction matters.

```python
counts = {}
for word in words:
    if word not in counts:
        counts[word] = 0
    counts[word] = counts[word] + 1

print(counts)
print(max(counts, key=counts.get))
```

### Question 8

Grouping with a dictionary. -> `{'c': ['cat', 'cow', 'crow'], 'd': ['dog', 'duck'], 'e': ['emu']}`, then `3`, then `c`.

IDENTICAL SHAPE TO Q7. The only differences are that the value starts as
`[]` rather than `0`, and you `append` rather than adding one.

Counting and grouping are the same loop with a different value type, and
between them they account for most of what people actually reach for a
dictionary to do. Once that clicks, grouping into SETS is the obvious third
variant — worksheet 07 Q4 does exactly that when it needs the distinct
categories each customer bought from.

```python
groups = {}
for animal in animals:
    letter = animal[0]
    if letter not in groups:
        groups[letter] = []        # a LIST this time, not a 0
    groups[letter].append(animal)

print(groups)
print(len(groups))
print(max(groups, key=lambda k: len(groups[k])))
```

### Question 9

Stretch — comprehensions, and an inversion that survives. -> `{'widget': 15, 'gizmo': 3, 'sprocket': 5}`, `{'widget': 30, 'gizmo': 6, 'sprocket': 10}`, `{3: 'red', 2: 'blue', 1: 'green'}`, then `3 3`.

THE LENGTHS MATCH THIS TIME — three in, three out — so nothing was lost.

That is not a property of the code, it is a fact about this particular
data: the three counts happened to be distinct. Compare worksheet 04 Q11,
where two equal values collided and the inverted dictionary came back a
pair short. The inversion code is character-for-character the same in both
cases.

The only thing that told you which situation you were in was comparing the
two lengths. So do it every time, including — especially — the times it
passes.

```python
in_stock = {k: v for k, v in stock.items() if v > 0}
print(in_stock)

doubled = {k: v * 2 for k, v in stock.items()}
print(doubled)

by_count = {v: k for k, v in counts.items()}
print(by_count)

print(len(counts), len(by_count))
```
