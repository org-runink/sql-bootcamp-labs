# Part 03 - List and Dictionaries Comprehensions

```python
squares = [x**2 for x in range(5)]
squares
```

**What you should see:** `[0, 1, 4, 9, 16]`. A list comprehension, `[expression for item in iterable]`, is a compact way to build a new list from an existing iterable, equivalent to writing a `for` loop that appends to an empty list.

```python
even_squares = [x**2 for x in range(10) if x % 2 == 0]
even_squares
```

**What you should see:** `[0, 4, 16, 36, 64]`. Adding `if x % 2 == 0` filters the comprehension down to only the even numbers in `range(10)` before squaring them.

```python
square_dict = {x: x**2 for x in range(5)}
square_dict
```

**What you should see:** `{0: 0, 1: 1, 2: 4, 3: 9, 4: 16}`. A dictionary comprehension works the same way as a list comprehension, but builds `key: value` pairs, `{key_expr: value_expr for item in iterable}`.

```python
even_square_dict = {x: x**2 for x in range(10) if x % 2 == 0}
even_square_dict
```

**What you should see:** `{0: 0, 2: 4, 4: 16, 6: 36, 8: 64}`, the even numbers from 0 to 9 mapped to their squares, combining a dictionary comprehension with an `if` filter.

```python
matrix = [[i*j for j in range(3)] for i in range(3)]
matrix
```

**What you should see:** `[[0, 0, 0], [0, 1, 2], [0, 2, 4]]`. A comprehension inside another comprehension builds a nested list (here, a 3x3 matrix), the outer comprehension builds each row, the inner one builds the values within that row.

```python
tuple_comp = tuple(x**2 for x in range(5))
tuple_comp
```

**What you should see:** `(0, 1, 4, 9, 16)`. **Important:** Python has no dedicated "tuple comprehension" syntax, `(x**2 for x in range(5))` on its own is actually a **generator expression** (it produces values lazily, one at a time, rather than building a list all at once). Wrapping it in `tuple(...)`, as done here, consumes that generator and collects its values into a tuple. This is a common point of confusion, `(...)` looks like it should work like `[...]` and `{...}` do for comprehensions, but it doesn't.

```python
set_comp = {x**2 for x in range(5)}
set_comp
```

**What you should see:** `{0, 1, 4, 9, 16}`. A set comprehension uses curly braces like a dictionary comprehension, but with a single expression instead of a `key: value` pair, `{expression for item in iterable}`, and (like any set) automatically drops duplicate values.
