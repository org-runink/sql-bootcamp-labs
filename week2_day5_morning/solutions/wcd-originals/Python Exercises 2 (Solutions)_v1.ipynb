### 3. Find the most expensive product

**HINT**
> - language: `python`
> - data: `products.csv`  
> - use `for loop` 
> - avoid built-in functions such as `max` and `min`

```python
with open('data/products.csv','r') as file:
    products = file.readlines()
```

```python
products[:10]
#Columns: product_id, product_category_id, product_name, product_description, product_price, product_image
```

```python
products_dict = {}
for x in products[1:]:
    name = x.split('\t')[2]
    price = float(x.split('\t')[-2])
    if name not in products_dict:
        products_dict[name] = price

products_dict
```

```python
max_price = 0
for x in products[1:]:
    if float(x.split('\t')[4]) > max_price:
        max_price = float(x.split('\t')[4])
```

```python
names = []
for x in products_dict:
    if products_dict[x] == max_price:
        names.append(x)
names
```

```python
# For each for loop, the memory usage and the speed is directly dependent on the length 
## Each of them has a memory usage of o(n), where n is the length of the object
```

```python
product = []
highest = 0
for row in products[1:]:
    price = float(row.split("\t")[4])
    name = row.split("\t")[2]
    if highest < price:
        product.append(name)
        highest = price       
product
```

```python
datetime.datetime.strptime(orders[1].split('\t')[1].strip('"'), "%Y-%m-%d %H:%M:%S").weekday()
```

```python
datetime.datetime.strptime('2021-10-08 00:00:00', "%Y-%m-%d %H:%M:%S").weekday()
# weekday starts with Monday as 0, and ends with Sunday as 6
```

```python
def check_friday(string):
    return datetime.datetime.strptime(string.strip('"'), "%Y-%m-%d %H:%M:%S").weekday() == 4
```

```python
check_friday('"2021-10-08 00:00:00"')
```

```python
sum([check_friday(x.split('\t')[1]) for x in orders[1:]])
```

### 7. Calculate number of sales order by state

**HINT**
> - language: `python`
> - data: `orders.csv` | `customers.csv`
> - use `dictionary` for state name lookups
> - requirement:
>   - open the customers.csv and build a dictionary that maps customer_id to state
>   - define a function that looks up state by customer_id and returns the sales orders and state

```python
with open('data/customers.csv','r') as file:
    customers = file.readlines()
```

```python
orders[:10]
```

```python
customers[:10]
```

```python
customer_dict = {}

for x in customers[1:]:
    customer_id = x.split('\t')[0]
    customer_state = x.split('\t')[-2].strip('"')
    customer_dict[customer_id] = customer_state
```

```python
def look_up(customer_id):
    return customer_dict[str(customer_id)]
```

```python
customer_dict
```

```python
look_up('10')
```

```python
counters = {}
for x in orders[1:]:
    cid = x.split('\t')[-2]
    state = look_up(cid)
    if state in counters:
        counters[state] += 1
    else:
        counters[state] = 1
```
