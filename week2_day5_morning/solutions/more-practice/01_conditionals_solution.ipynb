### Question 1

Seniority bands. -> `mid`, then `senior`, then `junior`.

3 falls through `>= 5` and lands on `>= 3`. Narrowest condition first,
widest last — worksheet 01 Q5 is what happens when you get that backwards.

Three runs, three branches. A band chain you have only run once has only
been tested once, and the branch that is wrong is always the one you did
not try.

```python
if years >= 5:
    print("senior")
elif years >= 3:
    print("mid")
else:
    print("junior")

years = 5
if years >= 5:
    print("senior")
elif years >= 3:
    print("mid")
else:
    print("junior")

years = 1
if years >= 5:
    print("senior")
elif years >= 3:
    print("mid")
else:
    print("junior")
```

### Question 2

`in` as a test. -> `True`, `True`, `True`, **`False`**, `True`, `False`.

`"N1" in region_codes` is `False` because `in` on a dictionary asks about
its **keys**. `N1` is a value, and `in region_codes.values()` is how you
ask that — a different question with a different cost, since values are
scanned one by one while keys are found instantly.

The last line is `False` because `"ana" in "Ana Lima"` is case-sensitive.
On a string, `in` is a substring test — `"na L" in "Ana Lima"` is `True`
too, which catches people who expect it to match whole words.

```python
print(role in allowed_roles)
print("manager" not in allowed_roles)
print("north" in region_codes)
print("N1" in region_codes)        # asks the KEYS, not the values
print("N1" in region_codes.values())
print("ana" in "Ana Lima")         # case-sensitive substring test
```

### Question 3

### Question 4

What `or` and `and` hand back. -> `guest`, `Ana`, an **empty line**, then `<class 'str'>`.

None of these produced `True` or `False`. `or` returns the **first truthy
operand**, or the last one if none are truthy; `and` returns the **first
falsy one**, or the last if all are truthy. The values come straight
through.

`supplied_name or fallback_name` is `"guest"` because `""` is falsy — and
that is the idiom this question exists for: `name = supplied or default` is
how you fill in a missing value in one line.

The blank third line is `supplied_name and username`, which is `""`. And
that is the trap: `""`, `0` and `[]` are all falsy, so `count = supplied or
10` quietly replaces a legitimate zero with 10. When zero is a real value,
test `is None` instead.

```python
print(supplied_name or fallback_name)
print(username or fallback_name)
print(supplied_name and username)
print(type(supplied_name or fallback_name))
```

### Question 5

Short-circuiting. -> `False`, `True`, `analyst`. **No `NameError`, twice.**

`False and anything` is already false, so Python never looks at the
right-hand side — including a name that does not exist. Same for
`True or anything`.

The third line is that behaviour doing a job: `len(allowed_roles) > 0 and
allowed_roles[0]` cannot raise `IndexError`, because the indexing is only
reached when the length check has already passed. That is the standard
way to guard an operation with its own precondition, and it is why the
order of the two halves matters — swap them and it raises.

```python
print(False and undefined_thing)   # never looks at the right-hand side
print(True or undefined_thing)     # already knows the answer

# The same trick, doing real work: the length check protects the indexing.
print(len(allowed_roles) > 0 and allowed_roles[0])
```

### Question 6

Comparing strings. -> `True`, **`True`**, **`True`**, `False`, `False`.

Strings compare character by character, by character code. `"Zoe" < "ana"`
is `True` because every capital letter sorts before every lowercase one —
`Z` is 90, `a` is 97. Lowercase both and it flips to `False`.

`"10" < "9"` is `True` for the same reason and is the more dangerous one:
it compares `"1"` with `"9"` and stops. As text, `"10"` really does come
before `"9"`, and every value read from a CSV is text. **Convert before
you compare**, or your sorted report puts 10 before 9.

```python
print("apple" < "banana")
print("Zoe" < "ana")
print("10" < "9")
print(10 < 9)
print("Zoe".lower() < "ana".lower())
```

### Question 7

Nested and flat, over three cases. -> `read only`, `full access`, `no access` — the nested and flat versions agreeing on all three.

Six lines of nesting, five of flat, same answers. At two conditions it is a
toss-up; at three the flat chain wins clearly, and the nested version is
still growing an indentation level per condition.

The loop over test cases is the part worth stealing. Three tuples and one
loop is how you check every branch of a rule, and it costs less than the
three copy-pasted blocks it replaces.

```python
for test_role, test_cert in [("analyst", False), ("analyst", True), ("intern", True)]:
    print(test_role, test_cert, "->")

    if test_role in allowed_roles:
        if test_cert:
            print("  nested:", "full access")
        else:
            print("  nested:", "read only")
    else:
        print("  nested:", "no access")

    if test_role in allowed_roles and test_cert:
        print("  flat:  ", "full access")
    elif test_role in allowed_roles:
        print("  flat:  ", "read only")
    else:
        print("  flat:  ", "no access")
```

### Question 8

Floats. -> `True`, `True`, **`False`**, `True`, then `reading 0.0 is under the 0.5 threshold`.

`0.1 + 0.1 + 0.1 == 0.3` is `False`: the sum is
0.30000000000000004. Not a Python bug — binary floating point cannot store
0.1 exactly, and the error survives the addition.

The fourth line is the fix. `abs(a - b) < tiny` asks "are these close
enough", which is the only sensible question to ask of two computed floats.
Worksheet 04 Q8 is the same problem inside a loop condition, where it costs
you forty extra passes.

`reading == 0.0` is `True` only because `0.0` was written down literally,
not calculated. Comparisons against exact literals are safe; comparisons
against arithmetic are not.

```python
print(reading < threshold)
print(reading == 0.0)
print(0.1 + 0.1 + 0.1 == 0.3)
print(abs((0.1 + 0.1 + 0.1) - 0.3) < 0.000001)

status = "over" if reading > threshold else "under"
print(f"reading {reading} is {status} the {threshold} threshold")
```

### Question 9

A missing key. -> `N1`, then `KeyError: 'east'`.

Square brackets on a dictionary demand that the key exists. `"east"` does
not, and Python raises rather than inventing a value.

Two safe forms, and they are not interchangeable:
- `if "east" in region_codes:` — check first, then index. Use it when a
  missing key means "do something different".
- `region_codes.get("east")` — returns `None` instead of raising, and
  `.get("east", "??")` returns a default of your choosing. Use it when a
  missing key has an obvious stand-in.

The one to avoid is catching the error and carrying on, which turns a
missing region into a silent gap.

```python
print(region_codes["north"])

# This is SUPPOSED to raise: KeyError. "east" is not a key.
#
# Two safe ways to ask:
#   if "east" in region_codes:   -- test first, then index
#   region_codes.get("east")     -- returns None instead of raising
print(region_codes["east"])
```
