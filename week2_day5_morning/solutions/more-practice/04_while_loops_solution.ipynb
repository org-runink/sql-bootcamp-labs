```python
# Extra practice 04 — While loops. Run this once.
tokens = ["get", "rows", "from", "orders", ";", "get", "more"]
letters = ["p", "y", "t", "h", "o", "n"]
word_chars = ["r", "o", "t", "a", "t", "o", "r"]

credits = 100
cost_per_call = 12

page_sizes = [3, 3, 3, 1, 0]      # what each API page returned; 0 means done
target = 1000

print(len(tokens), "tokens,", credits, "credits, cost", cost_per_call, "each")
```

### Question 1

The first five even numbers. -> `0 2 4 6 8`, then `stopped with n = 9`.

Two counters doing different jobs: `n` is the candidate and `found` is the
result count. Only `found` is in the condition, because the loop's job is
to collect five of something, not to reach a particular `n`.

`n` ends on 9, one past the last number printed — the same off-by-one as
worksheet 04 Q1. The increment sits at the bottom of the body **outside**
the `if`, so it runs on every pass, including the odd ones. Move it inside
and the loop stops advancing on the first odd number.

```python
n = 0
found = 0

while found < 5:
    if n % 2 == 0:
        print(n)
        found = found + 1
    n = n + 1

print("stopped with n =", n)
```

### Question 2

Spending a credit budget. -> `8 calls made, 4 credits left`.

8 × 12 = 96, leaving 4 — not enough for a ninth call, so the condition goes
false and the loop ends. `100 // 12` is 8 and gives the same answer
instantly, but only because every call costs the same; the loop still works
when they do not.

`credits >= cost_per_call` rather than `credits > 0` is the point. "Is
there anything left" and "is there enough for another one" are different
questions, and the second one is what stops this going negative.

```python
calls = 0

while credits >= cost_per_call:
    credits = credits - cost_per_call
    calls = calls + 1

print(calls, "calls made,", credits, "credits left")
```

### Question 3

Reading up to a terminator. -> four tokens, then `4 consumed, 3 left`.

The compound condition is doing two jobs: `i < len(tokens)` keeps the index
in range and `tokens[i] != ";"` stops at the terminator. Drop the first
and a stream with no `;` in it walks off the end — which is exactly Q9.

Order matters. `and` evaluates left to right and stops as soon as it knows
the answer, so the bounds check has to come **first** or the indexing
happens before it is safe. That is extra practice 01 Q5, doing real work.

`i` doubles as the count of what was consumed, which is why the leftover
arithmetic is a subtraction rather than a second counter.

```python
i = 0
while i < len(tokens) and tokens[i] != ";":
    print("token:", tokens[i])
    i = i + 1

print(i, "consumed,", len(tokens) - i, "left")
```

### Question 4

Two indexes meeting in the middle. -> `rotator` is `True in 3 comparisons`; `python` is `False in 1 comparisons`.

Seven letters, three comparisons — the two pointers each cover half the
word, and the middle letter is never compared with anything because it is
its own mirror. `left < right` (not `<=`) is what stops it comparing the
middle letter with itself and wasting a pass.

The second run does one comparison: `p` against `n` fails immediately and
`break` leaves. A search that can fail should fail as early as it can.

Two variables moving towards each other is a shape worth recognising —
it is the same idea behind binary search and behind merging two sorted
lists.

```python
left = 0
right = len(word_chars) - 1
same = True
checks = 0

while left < right:
    checks = checks + 1
    if word_chars[left] != word_chars[right]:
        same = False
        break
    left = left + 1
    right = right - 1

print(word_chars, "-- same backwards:", same, "in", checks, "comparisons")

left = 0
right = len(letters) - 1
same = True
checks = 0
while left < right:
    checks = checks + 1
    if letters[left] != letters[right]:
        same = False
        break
    left = left + 1
    right = right - 1

print(letters, "-- same backwards:", same, "in", checks, "comparisons")
```

### Question 5

A do-while, emulated. -> five `page N returned …` lines ending in `0`, then `10 rows in 5 requests`.

The body always runs at least once because the test is at the **bottom**.
That is what a do-while means, and Python has no keyword for it —
`while True:` with a `break` at the end of the body is the standard way.

It matters for paging: you cannot know whether there is a page until you
ask for one. A top-tested `while size > 0:` would need a fetch before the
loop as well as inside it, duplicating the line that does the work.

Five requests returned four pages of data and one empty one. The empty
request is not waste — it is the only thing that says there is no more.

```python
page = 0
rows = 0

while True:
    size = page_sizes[page]
    page = page + 1
    rows = rows + size
    print("page", page, "returned", size)
    if size == 0:
        break

print(rows, "rows in", page, "requests")
```

### Question 6

Doubling past a target. -> `1024 after 10 doublings`, `stopped by the guard: False`.

2¹⁰ is 1024, the first power of two past 1000. The loop overshoots by
design: the condition is checked between passes, so the value you end with
is always one step beyond the last time the condition held.

`stopped by the guard: False` is the line that matters. It says the loop
ended because it finished, not because the safety net caught it — and
printing that distinction is the difference between a guard that protects
you and a guard that hides a bug. Worksheet 04 Q2 is the same loop with
the answer the other way round.

```python
value = 1
doublings = 0

while value <= target and doublings < 50:
    value = value * 2
    doublings = doublings + 1

print(value, "after", doublings, "doublings")
print("stopped by the guard:", doublings == 50)
```

### Question 7

A while inside a while. -> three rows of four labels, then `inner body ran 12 times`.

3 × 4 = 12. The multiplication is the whole cost model of nesting, and it
is the same lesson as worksheet 02 Q12.

`col = 0` at the top of the outer body is the line this question exists
for. Put it above the outer loop instead and the inner condition is
already false on the second pass — you get one row of output, two silent
empty rows, and no error. A `for` loop cannot make this mistake, because
`range(4)` starts fresh every time.

```python
row = 0
inner_runs = 0

while row < 3:
    col = 0                    # reset on every outer pass
    line = ""
    while col < 4:
        line = line + f"{row},{col}  "
        inner_runs = inner_runs + 1
        col = col + 1
    print(line)
    row = row + 1

print("inner body ran", inner_runs, "times")
```

### Question 8

Building a word by draining a list. -> `p`, `py`, `pyt`, `pyth`, `pytho`, `python`, then `word: python` and `letters: []`.

`while letters:` uses the list's own truthiness, and `.pop(0)` guarantees
progress by making it shorter every pass. Those two facts together are why
this loop cannot run forever.

And `letters` is now empty. The loop consumed its input to produce its
output — fine here, a real problem if a later cell needs the list. Re-run
the setup cell to get it back.

That is the cost of `while list:` over `for item in list:`. The `for`
version reads the list without destroying it, and it is what you want
unless the consuming *is* the point, as in a work queue.

```python
word = ""
while letters:
    word = word + letters.pop(0)
    print(word)

print("word:", word)
print("letters:", letters)
```

### Question 9

`<=` against a length. -> all five sizes print, then `IndexError: list index out of range`.

A list of 5 items has indexes 0 to 4. `len()` is 5, so `page <= len(...)`
allows a sixth pass at index 5, and there is nothing there.

**When you compare an index with a length, the operator is `<`.** Never
`<=`. The same off-by-one hides in `range(len(x) + 1)` and in a slice
endpoint written one too far.

Note how it failed: every real value printed first, so the output looks
completely correct right up to the traceback. If this loop had been
appending to a list instead of printing, you would have had a full,
correct-looking result and an exception on top of it.

`for size in page_sizes:` cannot make this mistake at all, which is the
real answer.

```python
page = 0

while page <= len(page_sizes):
    print(page, page_sizes[page])
    page = page + 1

# This is SUPPOSED to raise: IndexError, on the pass where page == 5.
#
# A list of 5 items has indexes 0 to 4. len() is 5, so `page <= len(...)`
# allows a sixth pass that indexes off the end. The condition is `<`, never
# `<=`, whenever you are comparing an index with a length -- and this is why
# `for page_size in page_sizes:` is the better loop: it cannot be off by one.
print("never reached")
```
