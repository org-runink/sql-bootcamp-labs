PART A — if / else, and the operators that go inside

### Question 1

The rule from slide 9, written both ways. -> `Too expensive!` twice.

`3.49 > 2.99` is true, so the first version takes the `if` branch. The
second version asks `3.49 <= 2.99`, which is false, so it takes its `else`
branch — and that branch holds the same sentence. Same answer, because `>`
and `<=` are exact opposites and you swapped the branches to match.

If you swap the test but forget to swap the branches, you get the opposite
answer with no error at all. That is the single most common conditional
bug there is.

```python
if dozen_egg_price > 2.99:
    print("Too expensive!")
else:
    print("Just the right price!")

# Same rule, written from the other side. > and <= are exact opposites, so
# swapping the test AND the branches has to give the same answer.
if dozen_egg_price <= 2.99:
    print("Just the right price!")
else:
    print("Too expensive!")
```

### Question 2

The six comparison operators. -> `==` True, `!=` False, `>` False, `<` False, `>=` True, `<=` True.

`basket_total` is exactly `84.50`, so it is neither greater nor less — but
it is both `>=` and `<=`. That is worth holding onto: `>=` is not "a bit
more than", it is `>` **or** `==`.

Notice the setup cell printed `84.5`, not `84.50`. The trailing zero was
never stored; it was only in the way you typed it. Python prints the
shortest text that round-trips back to the same float.

```python
print("==", basket_total == 84.50)
print("!=", basket_total != 84.50)
print("> ", basket_total > 84.50)
print("< ", basket_total < 84.50)
print(">=", basket_total >= 84.50)
print("<=", basket_total <= 84.50)
```

### Question 3

Boolean operators. -> `True` (and), `True` (or), `False` (not), `True` (chained).

The basket is 84.50 and the membership is gold, so `and` is satisfied —
both halves are true. `or` only needs one, so it is true as well; it would
still be true on a £5 basket.

`not membership == "gold"` reads awkwardly but works, because `==` binds
tighter than `not`. `membership != "gold"` says the same thing better.

`0 <= fraud_score <= 1` is a genuine three-way test in Python. In most
languages that line is a bug: they evaluate `0 <= fraud_score` to a
boolean and then compare *that* with 1.

```python
print(basket_total >= 50 and membership == "gold")
print(basket_total >= 50 or membership == "gold")
print(not membership == "gold")

# Python lets you chain comparisons the way maths does. Most languages do not.
print(0 <= fraud_score <= 1)
```

PART B — elif chains, and the bug that hides in them

### Question 4

A correctly ordered grade chain. -> `C`, then `A`, then `F`.

77 falls through `>= 90` and `>= 80`, matches `>= 70`, and stops. 95
matches the first branch immediately. 12 matches nothing, so `else`
catches it.

Only ever ONE branch of an if/elif/else runs — the first true one. The
rest are not even evaluated. That is exactly what makes the next question
possible.

```python
score = 77
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("F")

# Re-run the same chain on the two ends. (No loop yet -- that is worksheet 02.)
score = 95
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("F")

score = 12
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("F")
```

### Question 5

The same bands in the wrong order. -> `C`. **A 95 is graded C, and nothing warns you.**

The chain stops at the first true test. Every score of 80 or 90 has
already passed `>= 70`, so the two `elif` branches below it are
unreachable — dead code that can never run for any input whatsoever.

No error, no warning, no crash. The program is doing precisely what you
wrote; what you wrote is not what you meant. **Order your bands from the
narrowest condition to the widest**, and the widest is what `else` is for.

```python
score = 95

if score >= 70:
    print("C")
elif score >= 80:
    print("B")
elif score >= 90:
    print("A")

# A 95 is graded "C". An if/elif chain stops at the FIRST true branch, and
# every score that reaches 80 or 90 has already passed the >= 70 test, so
# neither elif can ever run. Both are dead code. Nothing warns you.
```

PART C — Nesting, and flattening it again

### Question 6

Nested shipping rule. -> `15.0`.

`region` is `"north"` so the outer `if` is taken; `weight_kg` is 12.0 which
is over 10, so the inner `if` is taken. Two decisions, two levels of
indentation, four possible outcomes.

Each level of nesting doubles the number of paths you have to keep in your
head — and doubles the number you have to test.

```python
if region == "north":
    if weight_kg > 10:
        shipping = 15.0
    else:
        shipping = 8.0
else:
    if weight_kg > 10:
        shipping = 25.0
    else:
        shipping = 12.0

print(shipping)
```

### Question 7

The same rule flattened with `and`. -> `15.0 15.0`, then `True`.

Four branches, one level of indentation, and the same four outcomes. Note
that only the *first* branch needs `and`: by the time you reach the second
branch, `region == "north" and weight_kg > 10` has already been ruled out,
so "north" on its own is enough.

That is the payoff of an `elif` chain — every branch gets to assume that
everything above it was false, so the later tests can be shorter.

```python
if region == "north" and weight_kg > 10:
    shipping_flat = 15.0
elif region == "north":
    shipping_flat = 8.0
elif weight_kg > 10:
    shipping_flat = 25.0
else:
    shipping_flat = 12.0

print(shipping, shipping_flat)
print(shipping == shipping_flat)
```

PART D — Truthiness: what counts as True when it is not a comparison

### Question 8

Truthiness. -> `False`, `False`, `False`, `False`, then `True`, `True`, `True`. Then `nothing to do`.

Falsy: `0`, `0.0`, `""`, `[]` — and also `{}`, `set()`, `None`. Everything
else is truthy.

The last three are where people get caught. `"0"` is a string one character
long, not the number zero. `[0]` is a list holding one item, and a list
with something in it is truthy regardless of what that something is. `" "`
is a space — still a character.

So `if row:` and `if row[0]:` can disagree, and if your CSV parser hands
you the string `"0"` where you expected the number, `if value:` will read
it as true.

```python
print(bool(items_in_stock))    # 0
print(bool(customer_name))     # ""
print(bool(pending_orders))    # []
print(bool(0.0))
print(bool("0"))               # a one-character string
print(bool([0]))               # a list holding one item
print(bool(" "))               # a space is a character

if pending_orders:
    print("work to do")
else:
    print("nothing to do")
```

PART E — The conditional expression

### Question 9

The conditional expression. -> `False`, `False`, `True`. **The slide disagrees with Python here.**

`fraud_score` is 0.25, which is not over 0.85, so `fraud_flag` is `False`.
Slide 23 puts `True` under the one-liner and `False` under the equivalent
`if`/`else` — the same code cannot produce both. Python produces `False`
twice. Believe the interpreter.

The second half is the real lesson: `fraud_score > 0.85` is already a
boolean. `True if X else False` takes a bool and hands back the same bool
— nine extra characters that do nothing. Write the comparison.

```python
fraud_flag = True if fraud_score > 0.85 else False
print(fraud_flag)

# `fraud_score > 0.85` is ALREADY a bool. "True if X else False" just copies it.
fraud_flag2 = fraud_score > 0.85
print(fraud_flag2)
print(fraud_flag == fraud_flag2)
```

### Question 10

A conditional expression used as a value. -> `Eggs are expensive at 3.49`, printed twice.

This is the form worth keeping: the conditional expression **produces a
value**, so it can go anywhere a value goes — into a variable, into an
argument, into f-string braces.

Both lines print `3.49` and not `3.49` padded to two places, because the
f-string is showing the number, not a formatted price. `f"{price:.2f}"`
is how you would fix that.

The inline version is at the edge of readable. A second one on the same
line is over it.

```python
label = "expensive" if dozen_egg_price > 2.99 else "cheap"
print(f"Eggs are {label} at {dozen_egg_price}")

# The expression can live inside the braces. Readable here; stop before it
# needs a second one.
print(f"Eggs are {'expensive' if dozen_egg_price > 2.99 else 'cheap'} at {dozen_egg_price}")
```

### Question 11

Comparing text with a number. -> `True`, then `TypeError: '>' not supported between instances of 'str' and 'int'`.

The first line is fine: `84.5 > 50` compares a float with an int, and
Python converts between numeric types happily.

The second is not. `"84.50"` is text. Python refuses to guess whether you
wanted a numeric comparison or an alphabetical one, so it raises instead
of picking. `float("84.50") > 50` is the fix.

This matters more than it looks: everything read from a CSV, a form or a
JSON field arrives as text. **Convert at the boundary**, once, as the data
comes in — not scattered through every comparison downstream.

```python
print(basket_total > 50)      # a float against an int -- fine

# This is SUPPOSED to raise. Python will not guess what you meant by
# comparing text with a number; convert with float() first.
print("84.50" > 50)
```
