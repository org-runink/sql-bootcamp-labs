# Part 01 - Lists

## Creating Lists

````python
# Creating lists
numbers = [1, 2, 3, 4, 5]
fruits = ["apple", "banana", "orange"]
mixed_list = [1, "apple", True, 3.14]

# Empty list
empty_list = []

# Nested list
nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
````

**What you should see:** no output, this cell only creates variables. `numbers` and `fruits` hold a list of ints and a list of strings, `mixed_list` shows that a single Python list can hold different types at once, `empty_list` is a valid list with zero elements, and `nested_list` is a list of lists (a common way to represent a grid or matrix).

## Accessing Elements

````python
# Accessing elements
first_number = numbers[0]
second_fruit = fruits[1]
last_element = mixed_list[-1]

print("First Number:", first_number)
print("Second Fruit:", second_fruit)
print("Last Element:", last_element)
````

**What you should see:**
```
First Number: 1
Second Fruit: banana
Last Element: 3.14
```
List indices start at `0`, so `numbers[0]` is the first item. Negative indices count from the end, so `mixed_list[-1]` is the last item without needing to know the list's length.

## Working with Nested Lists

````python
# A list of lists: use two sets of square brackets to reach an inner element
print(nested_list)
print(nested_list[1])       # the second inner list
print(nested_list[1][2])    # the third element of the second inner list

# empty_list started with nothing, but can be filled in later
empty_list.append(100)
print(empty_list)
````

**What you should see:**
```
[[1, 2, 3], [4, 5, 6], [7, 8, 9]]
[4, 5, 6]
6
[100]
```
`nested_list[1]` selects the second inner list (`[4, 5, 6]`), and chaining a second index, `nested_list[1][2]`, reaches into that inner list to get its third element (`6`). This pattern, index-then-index, is how you navigate any list of lists (for example, a grid of rows and columns).

## Slicing

````python
# Slicing
some_numbers = numbers[1:4]  # Get elements at indices 1, 2, 3
first_two_fruits = fruits[:2]  # Get elements from start to index 1
last_two_fruits = fruits[-2:]  # Get elements from index -2 to the end

print("Some Numbers:", some_numbers)
print("First Two Fruits:", first_two_fruits)
print("Last Two Fruits:", last_two_fruits)
````

**What you should see:**
```
Some Numbers: [2, 3, 4]
First Two Fruits: ['apple', 'banana']
Last Two Fruits: ['banana', 'orange']
```
A slice `list[start:stop]` returns a *new* list containing elements from `start` up to (but not including) `stop`. Leaving `start` or `stop` empty means "from the beginning" or "through the end".

## Modifying Lists

````python
# Modifying lists
fruits[0] = "grape"  # Update element at index 0
fruits.append("kiwi")  # Add element to the end
fruits.insert(1, "pineapple")  # Insert element at index 1
fruits.extend(["blueberry", "watermelon"])  # Extend list with another list

print("Updated Fruits:", fruits)
````

**What you should see:** `Updated Fruits: ['grape', 'pineapple', 'banana', 'orange', 'kiwi', 'blueberry', 'watermelon']`. Lists are mutable, so `fruits[0] = ...` changes the list in place rather than creating a new one. `append()` adds one item to the end, `insert(i, item)` adds one item at a specific position, and `extend()` adds all the items from another iterable (as opposed to `append()`, which would add the whole list as one nested item).

## Removing Elements

````python
# Removing elements
fruits.remove("banana")  # Remove a specific element
popped_fruit = fruits.pop()  # Remove and return the last element
del fruits[0]  # Remove element at a specific index

print("Fruits after Removals:", fruits)
print("Popped Fruit:", popped_fruit)
````

**What you should see:** `Fruits after Removals: ['pineapple', 'orange', 'kiwi', 'blueberry']` and `Popped Fruit: watermelon`. `remove(value)` deletes the first matching value (raises `ValueError` if it isn't found), `pop()` removes and returns the last element (or the element at an index you pass in), and `del list[i]` deletes the element at a specific index without returning it.

## List Operations

````python
# List operations
all_fruits = fruits + ["orange", "apple"]  # Concatenation
repeated_fruits = fruits * 2  # Repetition
num_fruits = len(fruits)  # Length of the list

print("All Fruits:", all_fruits)
print("Repeated Fruits:", repeated_fruits)
print("Number of Fruits:", num_fruits)
````

**What you should see:**
```
All Fruits: ['pineapple', 'orange', 'kiwi', 'blueberry', 'orange', 'apple']
Repeated Fruits: ['pineapple', 'orange', 'kiwi', 'blueberry', 'pineapple', 'orange', 'kiwi', 'blueberry']
Number of Fruits: 4
```
`+` concatenates two lists into a new one, `*` repeats a list's contents, and `len()` returns the number of elements currently in the list.

## Other List Methods

````python
# Other list methods
numbers.sort()  # Sort the list
numbers.reverse()  # Reverse the list
index_of_apple = all_fruits.index("apple")  # Find the index of an element
count_apples = all_fruits.count("apple")  # Count occurrences of an element

print("Sorted Numbers:", numbers)
print("Index of Apple:", index_of_apple)
print("Count of Apples:", count_apples)
````

**What you should see:** `Sorted Numbers: [5, 4, 3, 2, 1]`, `Index of Apple: 5`, `Count of Apples: 1`. `sort()` and `reverse()` both modify the list in place (they return `None`, they don't return a new sorted list) here they run back to back, so the final order is descending. `index(value)` returns the position of the first match (raises `ValueError` if missing), and `count(value)` returns how many times it appears.

## Membership Testing (`in` / `not in`)

````python
# Check whether a value is (or isn't) present in a list
print("kiwi" in fruits)
print("mango" in fruits)
print("mango" not in fruits)
````

**What you should see:** `True`, `False`, `True`. `fruits` at this point is `['pineapple', 'orange', 'kiwi', 'blueberry']` (after the earlier updates and removals). `in` checks membership and returns a boolean, it's the standard, readable way to test for a value instead of manually looping through the list.

## Copying Lists: aliasing vs. a real copy

````python
# Assigning a list to a new variable name does NOT copy it,
# both names point to the same list in memory.
original = [1, 2, 3]
alias = original
alias.append(4)
print("original:", original)
print("alias:", alias)

# Use .copy() (or original[:] or list(original)) to get an independent list
true_copy = original.copy()
true_copy.append(5)
print("original:", original)
print("true_copy:", true_copy)
````

**What you should see:**
```
original: [1, 2, 3, 4]
alias: [1, 2, 3, 4]
original: [1, 2, 3, 4]
true_copy: [1, 2, 3, 4, 5]
```
`alias = original` doesn't create a new list, it just gives the same list a second name, so mutating `alias` also changes what `original` shows. This is a very common source of bugs. `original.copy()` (equivalently `original[:]` or `list(original)`) creates an independent list, so changes to `true_copy` afterward don't affect `original`.

## List Comprehensions

````python
# A concise way to build a new list by transforming or filtering another iterable
squares = [n ** 2 for n in range(1, 6)]
even_numbers = [n for n in range(1, 11) if n % 2 == 0]
uppercase_fruits = [fruit.upper() for fruit in fruits]

print("Squares:", squares)
print("Even Numbers:", even_numbers)
print("Uppercase Fruits:", uppercase_fruits)
````

**What you should see:**
```
Squares: [1, 4, 9, 16, 25]
Even Numbers: [2, 4, 6, 8, 10]
Uppercase Fruits: ['PINEAPPLE', 'ORANGE', 'KIWI', 'BLUEBERRY']
```
`[expression for item in iterable]` builds a new list by evaluating `expression` for every `item`. Adding `if condition` at the end filters which items are included. This is generally the more "Pythonic", and often faster, alternative to writing a `for` loop that repeatedly calls `.append()`.
