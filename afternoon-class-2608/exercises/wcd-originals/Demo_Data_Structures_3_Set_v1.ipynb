# Part 03 - Sets

## Creating Sets

```python
# Creating sets
letters = {'a', 'b', 'c', 'd', 'e'}
vowels = {'a', 'e', 'i', 'o', 'u'}

# Using the set() constructor
numbers_set = set([1, 2, 3, 4, 5])

# Sets automatically discard duplicate values
numbers_with_duplicates = set([1, 2, 2, 2, 2, 2, 2, 3, 4, 5])

x = set()
x.add(1)
x.add(2)
x.add(2)  # adding 2 again has no effect, it's already in the set
```

**What you should see:** no output, this cell only creates variables. `numbers_with_duplicates` collapses down to `{1, 2, 3, 4, 5}` even though the input list repeats `2` several times, sets never store duplicate values. `x` ends up as `{1, 2}` for the same reason, the second `x.add(2)` has no effect. Note that sets are also unordered, so printing a set may not show elements in the order you added them.

## Accessing Elements

```python
# Accessing elements (membership check)
print('a' in letters)  # Output: True
print('f' in letters)  # Output: False
```

**What you should see:** `True` then `False`. Sets don't support indexing like `letters[0]` (they're unordered, there is no "first" element), so the standard way to check for a value is the `in` operator, which is also very fast for sets.

## Set Operations

```python
# Set operations
union_set = letters.union(vowels)  # Union of two sets
intersection_set = letters.intersection(vowels)  # Intersection of two sets
difference_set = letters.difference(vowels)  # Elements in letters but not in vowels
symmetric_difference_set = letters.symmetric_difference(vowels)  # Elements in either set, but not both

print("Union Set:", union_set)
print("Intersection Set:", intersection_set)
print("Difference Set:", difference_set)
print("Symmetric Difference Set:", symmetric_difference_set)
```

**What you should see:** a `Union Set` containing all 8 unique letters from both sets, an `Intersection Set` of `{'a', 'e'}` (the letters common to both), a `Difference Set` of `{'b', 'c', 'd'}` (in `letters` but not `vowels`), and a `Symmetric Difference Set` of `{'b', 'c', 'd', 'i', 'o', 'u'}` (in exactly one of the two sets, not both). The exact order elements print in may vary since sets are unordered.

## Set Operations using Operators

```python
# Set operations using operators
all_letters = letters | {'f', 'g'}  # Union using |
common_letters = letters & {'a', 'b', 'c'}  # Intersection using &
remaining_letters = letters - {'a', 'b'}  # Difference using -
unique_to_either = letters ^ vowels  # Symmetric difference using ^

print("All Letters:", all_letters)
print("Common Letters:", common_letters)
print("Remaining Letters:", remaining_letters)
print("Unique to Either:", unique_to_either)
```

**What you should see:** the same results as the method calls above, just spelled differently: `|` is equivalent to `.union()`, `&` to `.intersection()`, `-` to `.difference()`, and `^` to `.symmetric_difference()`.

## Modifying Sets

```python
# Modifying sets
letters.add('f')  # Add a single element
letters.update({'g', 'h'})  # Add multiple elements
letters.remove('a')  # Remove a specific element
letters.discard('z')  # Remove if exists, else do nothing

print("Updated Letters:", letters)
```

**What you should see:** `Updated Letters: {...}` containing `b, c, d, e, f, g, h` (order may vary). `add()` inserts one element, `update()` inserts every element from another iterable, `remove()` deletes a value and raises `KeyError` if it's missing, while `discard()` deletes a value if present and does nothing if it isn't, that's why `discard('z')` doesn't raise an error even though `'z'` was never in the set.

## Removing Elements

```python
# Removing elements
letters.remove('b')  # Remove 'b' from the set
letters.discard('z')  # Try to remove 'z' (not present)

print("Letters after Removal:", letters)
```

**What you should see:** `Letters after Removal: {...}` containing `c, d, e, f, g, h` (`'b'` removed, order may vary). This reinforces the `remove()` vs `discard()` distinction from the cell above.

```python
letters.clear()
letters
```

**What you should see:** `set()`, the representation Python uses for an empty set. You can't write `{}` for an empty set, that syntax already means an empty dictionary, so `set()` is the only way to create or display one.

## Other Methods

```python
s1 = {1,2}
s2 = {1,2,4}
s3 = {5}
```

**What you should see:** no output, this cell only creates variables. `s1` is a subset of `s2` (every element of `s1` is in `s2`), while `s3` shares no elements with either.

```python
s1.isdisjoint(s2)
```

```python
s1.isdisjoint(s3)
```

**What you should see:** `False`, `True`, `True`, `True`. `isdisjoint()` returns `True` when two sets share no elements at all (`s1` and `s2` share `1` and `2`, so `False`, `s1` and `s3` share nothing, so `True`). `issubset()` checks whether every element of the calling set is also in the other set. `issuperset()` checks the reverse, whether the calling set contains every element of the other set.

```python
s1.issubset(s2)
```

```python
s2.issuperset(s1)
```

## Set Comprehensions

```python
# A concise way to build a new set by transforming or filtering another iterable
squared_evens = {n ** 2 for n in range(1, 8) if n % 2 == 0}
first_letters = {word[0] for word in ['apple', 'banana', 'avocado', 'blueberry']}

print("Squared Evens:", squared_evens)
print("First Letters:", first_letters)
```

**What you should see:** `Squared Evens: {4, 16, 36}` and `First Letters: {'a', 'b'}`. `{expression for item in iterable if condition}` works just like a list comprehension but builds a set, and since sets discard duplicates automatically, `first_letters` ends up with only the two unique starting letters even though four words went in.
