PART A — Creating and reading a tuple

### Question 1

Warm-up — the comma is what makes a tuple. -> `('salmon', 'mackerel', 'sardines')`, `(5,)`, `5`, `<class 'tuple'>`, `<class 'int'>`.

`(5,)` is a one-element tuple. `(5)` is just the integer 5 with a pair of
brackets round it, exactly like `(2 + 3)` — brackets group expressions
everywhere in Python, so they cannot also be the thing that creates a
tuple. The comma has to do that job.

This bites when a function expects a tuple and you hand it `(x)`. Nothing
raises; you simply passed the wrong type.

```python
oily_fish = ("salmon", "mackerel", "sardines")
single = (5,)        # the trailing comma is what makes this a tuple
not_a_tuple = (5)    # no comma -- these brackets just group an expression

print(oily_fish)
print(single)
print(not_a_tuple)
print(type(single))
print(type(not_a_tuple))
```

### Question 3

Slicing a tuple. -> `('red', 'green')`, `('green', 'blue')`, `('blue', 'green')`, `<class 'tuple'>`.

A slice of a tuple is a tuple, just as a slice of a list is a list. And
note this does not breach immutability: slicing never modifies anything,
it builds a new object out of an existing one.

```python
print(colors[:2])
print(colors[1:3])
print(colors[-2:])
print(type(colors[:2]))   # slicing a tuple returns a tuple
```

PART B — What a tuple will not let you do

### Question 5

Immutability, demonstrated. -> `TypeError: 'tuple' object does not support item assignment`.

THIS IS THE WHOLE DIFFERENCE between a tuple and a list. The identical
statement on a list — worksheet 01 Q5 — worked silently. Here it is an
error, raised at the line where you tried it rather than surfacing later
as mysteriously changed data.

That is the reason to choose a tuple: it does not merely happen to stay
unchanged, it cannot be changed. Anything holding a reference to it is
immune to the aliasing problem in worksheet 01 Q10.

```python
# This is SUPPOSED to raise. A tuple cannot be changed after it is built.
toronto[0] = 44.0
```

### Question 6

Rebuild instead of modify. -> `(44.0, -79.416667)`, then `(43.666667, -79.416667)` — the original, intact — then `(43.666667, -79.416667, 76.0)`.

You never changed a tuple here; you built new ones out of old ones, which
is always allowed. `toronto + (76.0,)` needs its trailing comma too, or
you would be asking Python to add a float to a tuple.

The cost is real: every "change" copies the whole thing. That is why a
list remains the right choice for a collection you expect to grow.

```python
toronto_moved = (44.0, toronto[1])   # a NEW tuple, built from the old one
print(toronto_moved)
print(toronto)                       # the original is untouched

trip = toronto + (76.0,)             # + builds a new tuple; note the comma
print(trip)
```

### Question 7

Unpacking. -> `43.666667 -79.416667`; `Data Ninja - Master's`; `first: red` and `rest:  ['green', 'blue', 'green']`.

The count must match: four names for four fields, or Python raises
`ValueError`. `*rest` lifts that restriction by absorbing whatever is
left — and note it produces a **list**, even though the source was a tuple.

Unpacking is why tuples are the natural shape for a record.
`lat, lon = point` says what the two numbers mean; `point[0]` and
`point[1]` leave you remembering which is which.

```python
lat, lon = toronto
print(lat, lon)

name, age, gender, education = user_profile   # names must match the count
print(name, "-", education)

first, *rest = colors   # * collects everything left over
print("first:", first)
print("rest: ", rest)
```

PART C — Operators, methods, and choosing between the two

### Question 8

Operators. -> `4`; `('red', 'green', 'blue', 'green', 'yellow', 'purple')`; `('red', 'green', 'blue', 'green', 'red', 'green', 'blue', 'green')`; `True`; `True`.

`+` and `*` both return NEW tuples and leave `colors` exactly as it was —
which is the only thing they could possibly do to an immutable object.
On lists the same operators also build new lists, so this behaviour is not
special to tuples; it is `append`/`extend` that are the odd ones out.

```python
print(len(user_profile))

print(colors + ("yellow", "purple"))   # a new, longer tuple
print(colors * 2)                      # a new, repeated tuple

print("green" in colors)
print("black" not in colors)
```

### Question 9

The only two methods. -> `1` and `2`.

`colors` holds two greens: `index` reports the first at position 1 while
`count` finds both — the same first-match rule as lists.

WHY THERE ARE ONLY TWO: `append`, `insert`, `remove`, `pop`, `sort`,
`reverse` and `clear` all modify the object in place. On an immutable type
there is nothing they could do, so they simply do not exist. `index` and
`count` survive because they only read. If you want a tuple sorted,
`sorted(my_tuple)` works fine — and hands you back a list.

```python
print(colors.index("green"))   # position of the FIRST green
print(colors.count("green"))   # but there are two of them

# index() and count() are the ONLY tuple methods, and both only READ.
# append/remove/sort/reverse would all have to modify the tuple in place,
# which is exactly what immutability forbids -- so they do not exist.
```

### Question 10

Stretch — unpacking inside a comprehension. -> `[43.65, 45.42, 49.28]` and `[(45.42, -75.7), (49.28, -123.12)]`.

`for lat, lon in location_history` unpacks each tuple as it loops, so you
can name the parts instead of indexing them. Compare
`[point[0] for point in location_history]` — same answer, but now the
reader has to know what index 0 means.

This pattern — iterate a list of tuples, unpack in the `for` — is the most
useful thing on this sheet. It is how worksheet 13 reads every one of its
order lines.

```python
latitudes = [lat for lat, lon in location_history]
print(latitudes)

northern = [(lat, lon) for lat, lon in location_history if lat > 44]
print(northern)
```
