PART A — Building and reading

### Question 1

Warm-up — three ways to build a tuple. -> `(0, 0)`, `('solo',)`, `(1, 2, 3)`, `<class 'tuple'>`.

Look at how `one_item` prints: `('solo',)`, comma included. Python shows
that comma deliberately, because it is the only thing distinguishing a
one-item tuple from a bracketed value. If you see a trailing comma in
output, it is telling you the type, not making a typo.

`tuple()` accepts any iterable, so `tuple("abc")` gives `('a', 'b', 'c')` —
the same character-by-character behaviour as `set("abc")`.

```python
origin = (0, 0)
one_item = ("solo",)          # trailing comma REQUIRED
from_list = tuple([1, 2, 3])  # any iterable can become a tuple

print(origin)
print(one_item)
print(from_list)
print(type(one_item))
```

Step slicing a tuple. -> `('Mon', 'Tue', 'Wed')`, `('Sat', 'Sun')`, `('Mon', 'Wed', 'Fri', 'Sun')`, `('Sun', 'Sat', 'Fri', 'Thu', 'Wed', 'Tue', 'Mon')`.

### Question 4

Unpacking, including `*` in the middle. -> `255 128 0`; `Ada studies Data Engineering`; then `Mon`, `['Tue', 'Wed', 'Thu', 'Fri', 'Sat']`, `Sun`.

`first, *rest, last` puts the star in the MIDDLE. The plain names on each
side are filled first, and `*rest` absorbs everything between them — so the
count still has to work out, with at least one item for every plain name.

Note `rest` came back as a LIST even though `week` is a tuple. That catches
people who then try to use a tuple method on it.

```python
red, green, blue = rgb
print(red, green, blue)

name, age, programme, gpa = student
print(f"{name} studies {programme}")

first, *rest, last = week    # * in the middle: takes whatever is left over
print(first)
print(rest)                  # a LIST, even though week is a tuple
print(last)
```

PART B — Two things worksheet 03 did not show you

### Question 6

Swapping without a temporary. -> `1 2`, then `2 1`.

`a, b = b, a` works because the RIGHT-hand side is evaluated first, building
the tuple `(2, 1)`, and only then is it unpacked into the names on the
left. Both old values are safely captured before either name is
reassigned — which is precisely what a temporary variable would have done
by hand.

This is probably the most-used tuple feature in Python, and most people who
reach for it have never noticed a tuple was involved at all.

```python
a = 1
b = 2
print(a, b)

a, b = b, a   # the right side becomes the tuple (2, 1), then unpacks
print(a, b)
```

### Question 7

A mutable item inside a tuple. -> `('fixed', [1, 2])`, then `('fixed', [1, 2, 3])`.

THIS DOES NOT CONTRADICT Q5. A tuple guarantees that its slots never
change, and slot 1 still points at the same list it always did. It makes no
promise whatsoever about that list's CONTENTS, and lists are mutable.

So `mixed[1].append(3)` is fine, while `mixed[1] = [9]` would still raise:
the first changes the list, the second would change the tuple.

The practical consequence is that "immutable" does not mean "safe to share"
unless everything inside is also immutable. It is also why a tuple like
this one cannot be used as a dictionary key — worksheet 07 Q9 needs the key
to be hashable all the way down.

```python
mixed = ("fixed", [1, 2])
print(mixed)

mixed[1].append(3)   # allowed -- we never reassigned mixed[1]
print(mixed)

# The tuple stores a reference to that list, and the reference has not
# changed -- it still points at the same list, which is all the tuple
# guarantees. The LIST is mutable, so its contents can change.
# `mixed[1] = [9]` would still raise, because that would replace the
# reference the tuple is holding.
```

### Question 9

Stretch — unpacking inside comprehensions. -> `['Mon', 'Tue', 'Wed', 'Thu']`, `['Tue', 'Thu']`, `83`.

`for day, temp in readings` names both halves as it loops, so the filter
reads `if temp > 19` instead of `if reading[1] > 19`. Across four rows that
is cosmetic; across a real record with eight fields it is the difference
between code you can read and code you have to decode.

This list-of-tuples shape — one tuple per record, inside a list that grows —
is exactly what worksheet 13 is built on.

```python
days = [day for day, temp in readings]
print(days)

warm_days = [day for day, temp in readings if temp > 19]
print(warm_days)

total = sum([temp for day, temp in readings])
print(total)
```
