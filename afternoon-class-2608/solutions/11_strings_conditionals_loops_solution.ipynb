### Question 2

sorted, negative index, slice. -> `[3, 5, 7, 8, 12, 16, 20]`, then `16`, then `[3, 5, 7]`, then `[8, 3, 12, 7, 20, 5, 16]` unchanged.

Sorting once into `ordered` and indexing it three times is the point.
Calling `sorted(numbers)` three separate times gives identical answers and
does the same work three times over.

`ordered[-2]` gives the second largest without needing the length. And the
final print confirms `sorted()` left the original alone — where `.sort()`
would have silently changed what every later question on this sheet sees.

````python
ordered = sorted(numbers)
print(ordered)
print(ordered[-2])   # second largest, counting from the end
print(ordered[:3])   # three smallest

print(numbers)       # untouched, because sorted() never modifies
````

### Question 3

Splitting on a two-character separator. -> `['Data engineering is mostly plumbing', 'The rest is naming things', 'Both are hard.']`, then `3`, then `DATA ENGINEERING IS MOSTLY PLUMBING`.

LOOK AT THE THIRD PIECE. `'Both are hard.'` still carries its full stop
while the first two lost theirs. That is not a bug: `split(". ")` removes
the separator it matched, and the final full stop is not followed by a
space, so it was never part of one.

Text splitting is full of exactly this. The rule that looks right on the
first two items is quietly wrong on the last, and only reading the output
tells you — the length, `3`, looks perfectly correct either way.

````python
parts = sentence.split(". ")
print(parts)
print(len(parts))
print(parts[0].upper())
````

### Question 4

Splitting twice. -> `ada@runink.org -> runink.org -> org`, `bo@example.co.uk -> example.co.uk -> uk`, `cai@data.example.com -> data.example.com -> com`.

`split("@")[1]` takes what follows the `@`. Then `split(".")[-1]` takes the
LAST dot-separated piece, which is why it copes with all three shapes at
once: one dot, two dots, and a `co.uk`.

Using `[1]` instead of `[-1]` on that second split would have returned
`org`, `co` and `example` — correct for the first address and wrong for the
other two. Counting from the end is what makes it general, and the
difference would not show up if you only tested the simple case.

````python
for address in emails:
    domain = address.split("@")[1]      # everything after the @
    tld = domain.split(".")[-1]         # the LAST dot-separated piece
    print(address, "->", domain, "->", tld)
````

### Question 5

Cleaning strings into numbers. -> `[1200.0, 4.5, 890.0, 22.0]` and a total of `2116.5`.

The replacements chain: `.replace("$", "")` runs first and `.replace(",", "")`
runs on its result. Each returns a NEW string and nothing is modified in
place — strings are immutable, exactly like tuples.

This is what "data cleaning" usually means in practice. The values arrived
as display text, and every one has to survive conversion before any
arithmetic is possible. Note `float("$1,200.00")` raises rather than
guessing, so the very first tag stops you until you deal with it.

````python
prices = []
for tag in price_tags:
    cleaned = tag.replace("$", "").replace(",", "")   # chain the replacements
    prices.append(float(cleaned))

print(prices)
print(round(sum(prices), 2))
````

PART B — Making decisions

### Question 6

if / elif / else. -> `95 high`, `73 medium`, `41 low`.

ORDER IS LOAD-BEARING. An `elif` is only reached when every test above it
has failed, which is why `score >= 60` never sees 95.

Write the branches the other way round — 60 first — and every score of 80
or more is labelled `medium`, with no error and no warning. The output
still looks like a plausible set of bands.

A chain of separate `if` statements rather than `elif` has the same bug,
and additionally sets `band` twice on the way through.

````python
for score in [95, 73, 41]:
    if score >= 80:
        band = "high"
    elif score >= 60:      # only reached when the first test failed
        band = "medium"
    else:
        band = "low"
    print(score, band)
````

### Question 7

`in` on a string. -> `a miss`, `y hit`, `z miss`, `t hit`.

`in` asks whether one string appears inside another, so no loop over
characters is needed — the same operator you used on lists, sets and dicts
in the earlier sheets, doing the analogous job here.

It matches substrings, not just single characters, so `"th" in "python"` is
also `True`. That is occasionally a surprise when the thing you are
checking turns out to be longer than one letter.

````python
for letter in ["a", "y", "z", "t"]:
    if letter in answer_word:
        print(letter, "hit")
    else:
        print(letter, "miss")
````

PART C — Repeating work

### Question 8

Accumulating with range. -> `55` and `110`.

`range(1, 11)` is 1 to 10, because the stop value is excluded — the same
rule as slicing. `range(2, 21, 2)` takes a step as its third argument and
needs `21` rather than `20` for exactly the same reason.

The accumulator pattern — start at 0, add inside the loop — is the one you
used to total revenue in worksheet 13, just without a dictionary to key it
by.

````python
total = 0
for n in range(1, 11):        # 1 to 10 -- the stop value is excluded
    total = total + n
print(total)

even_total = 0
for n in range(2, 21, 2):     # 2, 4, 6, ... 20
    even_total = even_total + n
print(even_total)
````

### Question 9

A while loop. -> `5`, `4`, `3`, `2`, `1` on separate lines, then `liftoff`.

`count = count - 1` is the entire difference between a countdown and a hung
kernel. A `for` loop advances its own counter for you; a `while` loop does
not, so whatever the condition depends on must be changed by the body.

Use `for` when you know how many passes you need, and `while` when you are
waiting for a condition to become false — as in Q11, where the number of
guesses is not known in advance.

````python
count = 5
while count > 0:
    print(count)
    count = count - 1     # WITHOUT this line the loop never ends
print("liftoff")
````

### Question 10

Three ways to filter. -> `[8, 12, 20, 16]` three times, then `True`.

Identical results — and note the even numbers keep their ORIGINAL order in
all three, because none of these sorts anything.

The comprehension is what most Python code uses. The loop is worth writing
once so you can see the comprehension is only a compressed form of it:
make a list, test each item, keep the survivors. `filter` is one to
recognise in other people's code rather than to reach for.

The last line chains two comparisons and is `True` only if both hold.

````python
evens_loop = []
for n in numbers:
    if n % 2 == 0:
        evens_loop.append(n)
print(evens_loop)

evens_comp = [n for n in numbers if n % 2 == 0]
print(evens_comp)

evens_filter = list(filter(lambda n: n % 2 == 0, numbers))
print(evens_filter)

print(evens_loop == evens_comp == evens_filter)
````

### Question 11

Stretch — Hangman, solved in 8 guesses with 3 lives left.

->
```
p hit  -> p_____
a miss -> p_____ | lives: 4
t hit  -> p_t___
e miss -> p_t___ | lives: 3
h hit  -> p_th__
o hit  -> p_tho_
n hit  -> p_thon
y hit  -> python
solved: python
```

TWO THINGS MAKE THIS WORK.

First, `board` is a LIST of one-character strings, not a string.
`board[position] = guess` is precisely the item assignment a string
refuses — worksheet 01 Q5 succeeded and worksheet 03 Q5 raised, and this is
where that distinction earns its keep. `"".join(board)` turns it back into
text only at the moment of printing.

Second, the inner loop reveals EVERY position matching the guess, not just
one. `answer_word.index(guess)` would have revealed only the first
occurrence — indistinguishable from correct on `python`, which has no
repeated letters, and quietly broken on a word like `letter`. Test data
without duplicates hides this class of bug completely.

The two `break` checks come after the guess is processed, so the winning
final letter is still displayed before the game ends.

````python
board = ["_"] * len(answer_word)   # a LIST, so positions can be assigned
lives = 5
guesses = ["p", "a", "t", "e", "h", "o", "n", "y"]

for guess in guesses:
    if guess in answer_word:
        for position in range(len(answer_word)):
            if answer_word[position] == guess:   # reveal EVERY occurrence
                board[position] = guess
        print(guess, "hit  ->", "".join(board))
    else:
        lives = lives - 1
        print(guess, "miss ->", "".join(board), "| lives:", lives)

    if "_" not in board:
        print("solved:", "".join(board))
        break
    if lives == 0:
        print("out of lives, the word was", answer_word)
        break
````
