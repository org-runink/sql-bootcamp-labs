### Create a dictionary named `person` with keys `"name"`, `"age"`, and `"city"`, and corresponding values `"John"`, `30`, and `"New York"`, respectively.

```python
person = {"name": "John", "age": 30, "city": "New York"}
person
```

### Create a dictionary named `car` using the `dict()` constructor with keys `"make"`, `"model"`, and `"year"`, and corresponding values `"Toyota"`, `"Camry"`, and `2018`, respectively.

```python
car = dict(make="Toyota", model="Camry", year=2018)
car
```

### Create a dictionary named `my_dict` using the `dict.fromkeys()` method with keys `"name"`, `"age"`, and `"city"`, and default value `"Unknown"`.

```python
my_dict = dict.fromkeys(["name", "age", "city"], "Unknown")
my_dict
```

### Access the value associated with the
- `"name"` key in the `person` dictionary and print it.
- `"make"` key in the `car` dictionary using the `get()` method to avoid `KeyError` and print it.

```python
print(person.get("name"))
print(car.get("make"))
```

### Attempt to access the value for a key that does not exist in the `person` dictionary (for example `"email"`) using square brackets, and explain the error you get. Then show how using `.get("email")` instead avoids that error.

```python
person["email"]
```

**What you should see:** accessing a key that doesn't exist with square-bracket notation (`person["email"]`) raises a `KeyError`, since `"email"` was never added to the `person` dictionary. This is why `.get()` is often the safer choice when a key might be missing — it returns `None` (or a default you specify) instead of crashing the program.

```python
print(person.get("email"))
```

### Loop over the `person` dictionary to print each key on its own line. Then loop over `person.items()` to print each key and its value together, on one line per pair.

```python
# Loop over keys only
for key in person:
    print(key)

print()

# Loop over key-value pairs together
for key, value in person.items():
    print(key, value)
```

### Add a new key-value pair `"occupation": "Engineer"` to the `person` dictionary.

```python
person["occupation"] = "Engineer"
person
```

### Update the value associated with the `"year"` key in the `car` dictionary to `2020`.

```python
car["year"] = 2020
car
```

### Remove the key-value pair with the key `"model"` from the `car` dictionary.

```python
del car["model"]
car
```

### Check if the key `"age"` exists in the `person` dictionary and print the result.

```python
print("age" in person)
```

### Get all keys from the `person` dictionary and print them.

```python
print(list(person.keys()))
```

### Get all values from the `person` dictionary and print them.

```python
print(list(person.values()))
```

### Get the number of key-value pairs in the `person` dictionary and print it.

```python
print(len(person))
```

### Use a dictionary comprehension to create a dictionary named `name_lengths` mapping each string in the list `["Alice", "Bob", "Charlie"]` to its length, and print it.

```python
name_lengths = {name: len(name) for name in ["Alice", "Bob", "Charlie"]}
name_lengths
```
