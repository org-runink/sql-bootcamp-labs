PART A — Creating a set, and what it silently throws away

### Question 1

Warm-up — three ways to build a set. -> `fruits` holds `apple`, `banana`, `orange` and has length `3`. `numbers_set` is `{1, 2, 3, 4, 5}` with length `5`, although seven values went in. `x` is `{1, 2}` with length `2`.

SET ORDER IS NOT GUARANTEED, and it genuinely differs between runs of the
same code — so the order yours print in will not always match what is
written here, and that is not a mistake. Compare contents, never
arrangement, and never write code that depends on the order.

The duplicates raised nothing: `set()` discarded them on the way in, and
the third `x.add(2)` did nothing at all. A set records presence, not
occurrences — if you need counts, that is a dictionary (worksheet 04).

```python
fruits = {"apple", "banana", "orange"}
numbers_set = set([1, 2, 2, 2, 3, 4, 5])   # duplicates collapse on the way in

x = set()      # NOT {} -- that would be an empty dict
x.add(1)
x.add(2)
x.add(2)       # adding something already present does nothing

print(fruits, len(fruits))
print(numbers_set, len(numbers_set))
print(x, len(x))
```

### Question 2

Membership, then the trap. -> `True`, `False`, then `TypeError: 'set' object is not subscriptable`.

`in` is the only way to look inside a set, and it is what sets are best at:
membership stays roughly instant however large the set grows, where the
same test on a list scans it item by item.

`letters[0]` fails because the question has no answer. A set stores no
order, so there is no "first" element to return — which is also why the
arrangement you saw last time you printed one guarantees nothing.

```python
print("a" in letters)
print("f" in letters)

# This is SUPPOSED to raise. A set has no order, so there is no "first" item
# for [0] to return.
print(letters[0])
```

### Question 3

Deduplicating a list. -> `7`, then a set of the 4 distinct names with length `4`, then those same 4 back as a list.

Seven signups, four distinct people. `set(...)` is the standard
three-character answer to "how many distinct X are there".

BUT YOU PAID FOR IT: the order people signed up in is gone, permanently,
and `list()` does not restore it — it just wraps whatever arrangement the
set happens to be in. If order matters, deduplicating through a set is the
wrong tool, and the loss is silent.

```python
print(len(signups))

unique_signups = set(signups)
print(unique_signups, len(unique_signups))

back_to_list = list(unique_signups)
print(back_to_list)
# 7 signups, 4 distinct people -- and the order they signed up in is gone.
```

PART B — The four set operations

### Question 4

The four operations, method form. -> union = all 7 skills; intersection = `linux` and `python`; difference = `bash`, `docker`, `terraform`; symmetric difference = those three plus `sql` and `statistics`, so 5 in total.

Read them as questions about two people. Union: what does the pair know
between them. Intersection: what do they both know. Difference: what can
the DevOps engineer do that the data scientist cannot. Symmetric
difference: where do they fail to overlap.

Intersection is the one you will reach for most — it is the set version of
an inner join, and it is how you answer "which of these also appear in
those" without a loop.

```python
print(devops_skills.union(data_skills))                 # either role
print(devops_skills.intersection(data_skills))          # both roles
print(devops_skills.difference(data_skills))            # devops only
print(devops_skills.symmetric_difference(data_skills))  # exactly one role
```

### Question 5

The operators agree with the methods. -> `True` four times.

`|` union, `&` intersection, `-` difference, `^` symmetric difference.
Same operations, shorter to write, and the form you will meet in other
people's code.

The methods do have one advantage: `.union(other)` accepts any iterable —
a list, a tuple, a string — while `|` insists both sides are already sets.

```python
print((devops_skills | data_skills) == devops_skills.union(data_skills))
print((devops_skills & data_skills) == devops_skills.intersection(data_skills))
print((devops_skills - data_skills) == devops_skills.difference(data_skills))
print((devops_skills ^ data_skills) == devops_skills.symmetric_difference(data_skills))

# | union   & intersection   - difference   ^ symmetric difference
# The brackets are not required -- | binds tighter than == -- but they make
# the intent obvious, which matters more.
```

### Question 6

Difference is not symmetric. -> `{bash, docker, terraform}`, then `{sql, statistics}`, then `False`.

Union and intersection give the same answer whichever side you write
first. Difference does not: `a - b` asks what `a` has that `b` lacks, and
that is a genuinely different question from `b - a`.

Get the operands backwards and you produce a perfectly plausible set that
answers the wrong question, with nothing to tell you. When you actually
want both answers at once, that is precisely what `^` is: `(a - b) | (b - a)`.

```python
print(devops_skills - data_skills)   # in devops, not in data
print(data_skills - devops_skills)   # in data, not in devops
print((devops_skills - data_skills) == (data_skills - devops_skills))

# Difference is not symmetric: "what have I got that you haven't" is a
# different question from "what have you got that I haven't". Use ^ when you
# want both answers at once.
```

PART C — Changing a set, comparing sets, and a stretch

### Question 7

Changing a set. -> after `add`, 6 letters (a–f); after `update`, 8 (a–h); after `remove("a")`, 7 (b–h); `discard("z")` changes nothing and it is still 7. The copy prints `set()` while `letters` keeps all 7.

`discard("z")` was a no-op and nothing complained — that is the whole
difference from `remove`, which Q8 shows.

Note the empty set prints as `set()`, not `{}`. `{}` is how an empty
dictionary prints, which is the same reason you cannot create an empty set
with it.

And `clear()` on the copy left `letters` alone, because `copy()` made a
real second set. The worksheet 01 Q10 aliasing lesson applies to every
mutable structure, sets included.

```python
letters.add("f")            # one item
print(letters)

letters.update({"g", "h"})  # every item of another collection
print(letters)

letters.remove("a")         # a really is there
print(letters)

letters.discard("z")        # "z" is NOT there, and nothing happens
print(letters)

spare = letters.copy()
spare.clear()
print(spare, letters)
```

### Question 8

remove vs discard. -> `discard("nope")` prints the set unchanged; `remove("nope")` raises `KeyError: 'nope'`.

Same job, opposite attitude to being wrong. `discard` means "make sure
this is not in the set, I do not care whether it was". `remove` means
"this should be here, and if it is not then something upstream is broken
and I want to know now".

Choose deliberately. Reaching for `discard` to silence an error you have
not understood is how a real bug gets buried.

```python
letters.discard("nope")   # not present, and discard does not care
print(letters)

# This is SUPPOSED to raise. remove() insists the item exists.
letters.remove("nope")
```

### Question 9

Comparing whole sets. -> `True`, `True`, `True`.

`isdisjoint` asks whether the overlap is empty — clearer and cheaper than
`len(a & b) == 0`, because it can stop at the first item in common.

`issubset` and `issuperset` are one relationship read from opposite ends:
`{'a','e'}.issubset(vowels)` and `vowels.issuperset({'a','e'})` ask the
same question and must always agree.

```python
print(vowels.isdisjoint({"x", "y"}))    # no overlap at all?
print({"a", "e"}.issubset(vowels))      # is every one of mine in yours?
print(vowels.issuperset({"a", "e"}))    # does mine contain every one of yours?

# The last two are the same question asked from opposite sides.
```

### Question 10

Stretch — set comprehensions. -> `initials` is `{a, b, c, d}` with length 4; `long_names` is `{ana, cai, dee}` with length 3.

`signups` has 7 entries, so the comprehension produced 7 initials — and
the `{}` kept 4. You never wrote a deduplication step: choosing `{}` over
`[]` WAS the deduplication.

`long_names` is doing two reductions at once, which is worth separating in
your head because only one of them is visible in the code. The `if`
dropped `bo` for being two characters; the set silently collapsed the
three `ana` entries into one.

```python
initials = {name[0] for name in signups}
print(initials)

long_names = {name for name in signups if len(name) > 2}
print(long_names)

# signups has 7 entries but initials has 4: the comprehension produced 7
# letters and the set kept only the distinct ones. Deduplication is not a
# step you perform, it is what the {} does.
```
