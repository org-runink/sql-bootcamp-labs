# Week 4, day 3 (morning) — Worksheet 04 SOLUTIONS: Frameworks, regulation, and the DAMA wheel

Governance is increasingly not optional. This sheet covers the regulatory
picture and the most widely referenced framework.

> *Discussion, not coding.*

**These are discussion answers, not a marking scheme.** Where a question asks for a judgement, the reasoning matters more than matching what is written here — a group that disagrees well has done the exercise.

STEP 1 — the regulatory map

### Question 1

The lecture lists GDPR (EU), CCPA and CPRA (California), PIPEDA (Canada),
PIPL (China), POPIA (South Africa) — and notes this is a global trend.

Your company operates in three of those jurisdictions. Name **two engineering
consequences** — things you would have to build or change — that follow from
that fact alone.

The strong answers are architectural, not legal:

**Data residency and regional isolation.** Some regimes require data to stay in
a jurisdiction. That is a storage-topology decision — separate regions, separate
keys, sometimes a separate operator entirely. The lecture's Azure examples make
this concrete: US Gov regions for FedRAMP, China operated under Chinese law,
Germany's data-trustee model.

**Deletion and subject-access as a feature.** "Delete everything about this
person" has to be executable across every copy, including the warehouse,
backups, and any derived model. If you cannot enumerate where a person's data
went, you cannot comply — which makes lineage a compliance control, not a
convenience.

Groups often add: consent tracking as a column, retention policies as
automation, and audit logs that prove who accessed what.

STEP 2 — the DAMA wheel

### Question 2

DAMA-DMBOK names ten knowledge areas: data architecture; modeling and
design; storage and operations; security; integration and interoperability;
document and content management; reference and master data; data warehousing and
BI; metadata management; data quality.

Pick the **three** you have personally touched in this course so far, and name
where.

Almost everything this course has done sits in three or four of them:

**Data modeling and design** — the star schema, the grain discussion, the Type 6
dimension. Conceptual, logical and physical models are exactly this area.

**Data warehousing and BI** — the layered build, raw through marts, and the
reports at the end.

**Data quality** — the dbt tests: `not_null`, `unique`, `relationships`,
`accepted_values`, plus the custom range and row-count tests.

**Metadata management** — descriptions, doc blocks and lineage in the dbt
project. Often the one students have touched without realising it had a name.

The takeaway the lecture draws: trusted data comes from many connected
disciplines, not one activity. No single team "does governance".

STEP 3 — people, process, technology

### Question 3

Governance is described as a people, process **and** technology problem.
Take one failure from the Worksheet 01 story — a stale copy, an unencrypted
copy, or an inherited bias — and write the fix at all three levels.

Take the **stale copy**:

*People.* Someone owns Dataset B and is accountable for announcing changes.
Consumers are named, not anonymous. Without this the other two levels have
nobody to trigger them.

*Process.* A change to a shared dataset requires notifying registered consumers,
with a deprecation window. New consumers register rather than copy.

*Technology.* Access is granted to the source rather than by handing out copies;
freshness is monitored and alerts on staleness; lineage shows who is downstream.

The lesson is that the technology fix alone does not work. You can build
freshness monitoring and still have five stale copies, because the copies were a
*people and process* failure. This is why "buy a catalogue" rarely fixes
governance.

STEP 4 — where to start

### Question 4

You have just joined an organisation with no governance programme. You get
**one** initiative in your first quarter. What is it, and why that one before
the others?

There is no single right answer, but the reasoning separates strong from
weak. Common defensible choices:

**An inventory with owners.** You cannot govern what you cannot list, and
"who owns this?" unblocks every later conversation. Cheap, unglamorous, high
leverage.

**Classification of sensitive data.** If the regulatory exposure is real, this
is the one where being wrong is most expensive. Do it first when the risk is
legal rather than analytical.

**Definitions for the top ten metrics.** Choose this when the pain is that
reports disagree, because it converts an endless argument into a decision.

The weak answer is "implement a governance tool". A catalogue populated by
nobody, with no owners named, is a more expensive version of the same problem.
