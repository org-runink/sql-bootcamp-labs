# Week 4, day 3 (morning) — Worksheet 05 SOLUTIONS: Case study — growth through acquisition

A design activity on the lecture's case study.

> *This is a design exercise. There is nothing to run.*

**The company.** Defence industry. Five years old, $5B revenue, ten recent
acquisitions, multiple subsidiaries, many disconnected data systems.

**Time:** 30–40 minutes including discussion. Work in groups.

**These are discussion answers, not a marking scheme.** Where a question asks for a judgement, the reasoning matters more than matching what is written here — a group that disagrees well has done the exercise.

STEP 1 — name the failure modes

### Question 1

The lecture names four challenges: data silos, inconsistent reference data,
no trusted central platform, and innovation without governance. For **each**,
write one sentence on how it shows up day to day for someone trying to do their
job.

**Data silos.** Each of the ten subsidiaries manages its own data, so a
question spanning two of them takes weeks: data is hard to combine, definitions
are inconsistent, reports do not match, teams duplicate work, and business users
cannot find trusted data.

**Inconsistent reference data.** The same supplier, part or cost centre has a
different code in each subsidiary. Any consolidated number requires a mapping
somebody maintains by hand, and that mapping is itself ungoverned.

**No trusted central platform.** There is nowhere authoritative to look, so
people build their own extracts — recreating Worksheet 01's story at company
scale.

**Innovation without governance.** Teams adopt AI tools with no policy, so they
either leak sensitive data into unapproved services, or get blocked entirely by
a nervous security team. Both outcomes are bad; the second is often invisible.

STEP 3 — sequence the work

### Question 2

You cannot fix all four at once. Put them in the order you would tackle
them and justify the sequence — specifically, say what each step **unblocks**.

A defensible sequence, with the justification carrying the weight:

**1. Reference data.** Least glamorous, most unblocking. Without shared codes
for supplier, part and cost centre, nothing can be consolidated — so every later
step produces numbers people distrust.

**2. A central platform for the highest-value domain.** Not all ten subsidiaries
at once; one domain, done properly, that proves the pattern and gives the
organisation something trustworthy to point at.

**3. Silos.** With codes and a platform, integration becomes mechanical rather
than political. Attempting it first means negotiating ten definitions
simultaneously.

**4. AI policy.** Needs the first three to be meaningful — a policy about which
data may be used presupposes knowing what data exists and who owns it. That
said, an interim guardrail is worth issuing on day one, because the risk is live
whether or not you are ready.

Groups that put AI policy first usually argue risk exposure, which is
legitimate — the distinction is between a stopgap rule and the real programme.

STEP 4 — the operating model

### Question 3

Sketch who does what. Name the **roles** you would create or assign — not
people, roles — and for each say what decision they own. Three to five roles.

A workable shape:

**Data owner** (per domain, in the business). Owns the definition, decides who
gets access, and is accountable for quality. Must be someone with authority in
the business, not a technologist.

**Data steward** (per domain, hands-on). Maintains the definitions, reference
codes and quality rules day to day. Does the work the owner is accountable for.

**Platform / data engineering.** Owns the mechanisms — pipelines, tests,
lineage, access enforcement. Implements decisions; does not make them.

**Governance council** (cross-domain). Resolves conflicts between domains and
sets standards. Exists because "active customer" arguments cannot be settled
inside one domain.

**Security / privacy.** Owns classification and regulatory interpretation.

The distinction to press on: owners *decide*, engineers *enforce*. Groups that
put every decision with the data team have recreated the problem — the team with
the least business context making the most business-loaded calls.

STEP 5 — what you would measure

### Question 4

A year in, how would you know the programme worked? Name **three
measures**, and for each say how it could be gamed.

Good measures are outcomes; every one of them is gameable, which is worth
saying out loud.

**Time to answer a cross-subsidiary question.** *Gamed by* asking easier
questions, or by pre-building the specific reports being timed.

**Share of reports built on certified datasets.** *Gamed by* certifying
everything, which makes the label meaningless.

**Number of disputed numbers in exec meetings.** *Gamed by* people quietly
stopping raising disputes — the metric improves while trust falls.

Weak answers count artefacts: datasets catalogued, policies written, tools
deployed. Those measure activity, not whether anyone trusts the data — and they
are the easiest of all to game.
