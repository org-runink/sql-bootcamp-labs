# Week 4, day 3 (morning) — Worksheet 03 SOLUTIONS: Data quality, trust, and AI readiness

AI systems learn from historical data. If the data carries bias, missing
context or unfair patterns, the model reproduces them — and the problem is
usually the data, not the algorithm.

> *Discussion, not coding. One question here is an ethics judgement with no
> single right answer; the disagreement is the point.*

**These are discussion answers, not a marking scheme.** Where a question asks for a judgement, the reasoning matters more than matching what is written here — a group that disagrees well has done the exercise.

STEP 1 — the model is only as good as its examples

### Question 1

The lecture walks through a classifier improving as its training data
improves — more examples, more *representative* examples, then correctly
labelled examples. For each of those three improvements, name the governance
activity that would produce it in a real organisation.

**More examples** — data access. The blocker is usually not collection but
permission: 68% of respondents in the lecture's survey said poor access to
high-quality data slows AI adoption. Governance decides who may use which
dataset for what purpose, and a clear decision unblocks far more than a vague
one.

**More representative examples** — data quality and lineage. Knowing *what
population* the data covers requires knowing where it came from and what was
filtered on the way. Representativeness is a lineage question before it is a
statistics question.

**Correct labels** — data quality and stewardship. Labels are usually produced
by people, and their accuracy depends on a definition someone owns and
maintains. This is master/reference data management wearing a different hat.

STEP 2 — an ethics judgement

### Question 2

A company wants to predict which employees are at high risk of burnout, so
it can offer support and reduce turnover. It proposes collecting: laptop usage
(times, private/business, keystrokes), telephone usage, calendar entries, email
behaviour, and personal information (age, gender, marital status).

**Should the company do this? Under what conditions?** Take a position as a
group and defend it.

There is no single right answer, and a group that splits is doing it right.
What matters is the *conditions* named. Strong answers converge on:

**Purpose limitation.** The stated purpose is employee wellbeing. Would the same
data be used for performance review, promotion, or redundancy selection? If it
could be, it eventually will be — so the constraint has to be structural, not a
promise.

**Consent that is real.** Consent given to an employer is rarely free; there is
a power imbalance. Opt-in with no consequence for opting out is a much stronger
test than a policy everyone clicks through.

**Proportionality.** Keystroke logging and private-call metadata are
extraordinarily invasive for the stated benefit. Would aggregate, anonymised
signals achieve most of it?

**Sensitive attributes.** Age, gender and marital status are protected
characteristics. A model trained on them will encode them, and "we only used it
to predict burnout" is not a defence if the output influences employment.

**Accountability.** Who is answerable when the model is wrong about someone?

The lecture's framing applies: the issue is rarely the algorithm. It is the
data, the bias, and insufficient governance over how the model was trained and
evaluated.

STEP 3 — AI readiness is data readiness

### Question 3

Nearly **68%** of surveyed organisations said poor access to high-quality
data slows AI adoption. If that is true, an organisation wanting to "do AI"
should probably fix something else first. Name **three things** you would fix
before training anything, in priority order, and justify the order.

Reasonable orderings differ; the justification matters more than the list.
A defensible one:

**1. Access.** If teams cannot legally or practically reach the data, nothing
else matters. This is a governance decision, and it is often the cheapest to fix
— frequently it is unclear rather than forbidden.

**2. Quality and definition.** Once reachable, the data has to mean something
stable. Inconsistent definitions and unmeasured quality produce models that are
confidently wrong, which is worse than no model.

**3. Lineage and documentation.** Needed to answer "what data was used?" — which
you will be asked the first time a prediction is disputed, and which regulation
increasingly requires.

The through-line: none of these are model work. They are governance work that
happens to be the prerequisite.

STEP 4 — the five questions

### Question 4

For AI use cases the lecture says governance helps teams ask: *What data was
used? Is it representative? Is sensitive information protected? Are outputs
monitored? Who is accountable?*

Pick any AI use case — yours or one of the lecture's (supply chain optimisation,
vendor scoring, demand forecasting, quality inspection). Answer all five for it.

Take **procurement vendor scoring**:

*What data was used?* Historical purchase orders, delivery timeliness, defect
rates, invoice disputes — plus, dangerously, any free-text notes buyers wrote.

*Is it representative?* Vendors used more often generate more data, so
established suppliers look better-measured than new ones. The model may encode
incumbency as quality.

*Is sensitive information protected?* Vendor pricing is commercially sensitive
and often contractually confidential. Free-text notes may name individuals.

*Are outputs monitored?* A score that quietly excludes a vendor from
consideration needs review, or the feedback loop is invisible — the vendor never
gets work, so never generates data to improve its score.

*Who is accountable?* Procurement owns the decision; the data team owns the
score's construction. Both must be named, or "the model said so" becomes the
answer.
