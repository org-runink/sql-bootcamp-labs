# Week 4, day 3 (morning) — Worksheet 02 SOLUTIONS: Governance, management, and where they live

Two words that get used interchangeably and should not be, and where each one
actually shows up in a data platform.

> *Discussion, not coding.*

**These are discussion answers, not a marking scheme.** Where a question asks for a judgement, the reasoning matters more than matching what is written here — a group that disagrees well has done the exercise.

STEP 1 — two words

### Question 1

**Data governance** and **data management** are not the same thing. Write a
one-sentence distinction, then give one concrete example of each from any
system you have worked with.

**Data management** is the practice of overseeing data as an enterprise
asset — the *doing*. DAMA lists the disciplines: architecture, modeling,
storage and operations, security, integration, content management, master and
reference data, warehousing and BI, metadata, and quality.

**Data governance** is the authority over that work — the *deciding*. Who owns
this dataset, who may see it, what it means, what quality bar it must meet, and
who is accountable when it is wrong.

Put crudely: management builds and runs the pipeline; governance decides what
the pipeline is allowed to do and who answers for it. A team can be excellent at
management and have no governance at all — that is exactly the state Department
B's data ended up in.

STEP 2 — why an engineer should care

### Question 2

The lecture claims governance matters to data engineers specifically,
because engineers build the systems everyone else relies on. Pick **three** of
these properties and, for each, name something *you* would build or change to
support it: trusted, secure, documented, discoverable, consistent, compliant,
AI-ready.

The point is that each property has an engineering surface:

**Documented** — column descriptions and doc blocks that ship with the model,
not in a wiki that rots. You saw this in the dbt project: `_stg__docs.md` and
the `description:` fields.

**Trusted** — tests that fail the build. A `not_null` or `relationships` test is
governance expressed as code: it encodes an agreement about what the data must
look like, and enforces it on every run.

**Discoverable** — lineage. If a model declares its inputs, the platform can
answer "where did this come from" without anyone maintaining a diagram.

**Consistent** — one definition, referenced everywhere. A metric defined once in
a semantic layer cannot drift between dashboards.

**Secure / compliant** — access granted by role, and sensitive columns
classified. Mostly platform features, but engineers decide the schema layout
that makes them grantable.

Without governance, a warehouse becomes a collection of tables nobody fully
understands or trusts. That is a failure engineers can cause and engineers can
prevent.

STEP 3 — where governance appears

### Question 3

Governance shows up across the whole lifecycle: source systems, ingestion,
storage, transformation, analytics models, BI dashboards, AI/ML training sets,
and access workflows. Pick **two adjacent stages** and describe what governance
looks like at each, and what breaks if it is missing at the boundary between
them.

A good answer treats the boundary as the risky part.

*Example — ingestion → storage.* At ingestion, governance means knowing the
source, its owner, and its refresh contract. At storage, it means the landing
zone is access-controlled and classified. If the boundary is ungoverned, data
lands with no record of where it came from — and six months later nobody can say
whether a column is authoritative or a one-off backfill.

*Example — transformation → BI.* At transformation, governance is tests,
definitions and lineage. At BI, it is certified datasets and clear ownership of
each dashboard. If the boundary is ungoverned, analysts build on whatever table
they found, and a change upstream silently breaks a dashboard nobody knew
existed. That is what an **exposure** declares.

STEP 4 — inconsistent definitions

### Question 4

The lecture flags *inconsistent business definitions* as a core problem.
Take one term — "active customer", "revenue", "churn", or one from your own
work. Write **two** plausible definitions that different teams might use, then
say what breaks when both are in play.

Take **active customer**. Marketing: anyone who opened an email in 90 days.
Finance: anyone with a paid invoice this quarter. Both are reasonable; both are
defensible; they produce different numbers.

What breaks is not the number — it is *trust*. Two reports disagree, both are
correct under their own definition, and the meeting spends its time
reconciling instead of deciding. Worse, nobody can tell whether a difference is
a definition mismatch or a genuine data error, so real errors hide inside the
noise.

The fix is not picking a winner — it is naming them. `active_customer_marketing`
and `active_customer_finance`, each defined once, documented, and used
consistently. That is a governance decision that an engineer then encodes.
