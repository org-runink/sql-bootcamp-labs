> *For most beginner/intermediate warehouse projects, start with a star schema
> unless there is a clear reason to snowflake the dimensions.* — L03, slide 21

Slide 21 compares the three patterns in a table:

| Model | Key idea | Best use case | Main trade-off |
|---|---|---|---|
| Star | fact table connected directly to dimension tables | simple reporting and dashboards | may duplicate dimension details |
| Snowflake | some dimensions split into sub-dimensions | reusable hierarchies, cleaner management | more joins, slightly harder queries |
| Galaxy | multiple fact tables share common dimensions | comparing multiple business processes | more complex model design |

"May duplicate dimension details" and "more joins" are both real costs, and both
are measurable. This worksheet builds all three shapes on the enrollment data and
puts numbers on the trade-offs instead of adjectives.
