## Worked example: retail sales

**Industry:** Retail.
**Business process:** A customer buys products at a store checkout.
**Grain:** One row per product sold on a single sales transaction line.

**Fact table: `FACT_SALES`**

| Column | Role | Notes |
|---|---|---|
| `DATE_KEY` | dimension key | links to the date dimension |
| `PRODUCT_KEY` | dimension key | links to the product dimension |
| `STORE_KEY` | dimension key | links to the store dimension |
| `CUSTOMER_KEY` | dimension key | links to the customer dimension |
| `QUANTITY_SOLD` | measure | units sold on the line |
| `SALES_AMOUNT` | measure | line revenue |
| `DISCOUNT_AMOUNT` | measure | discount applied to the line |

**Dimension tables**

| Dimension | Example attributes |
|---|---|
| `DIM_DATE` | date, day of week, month, quarter, year, holiday flag |
| `DIM_PRODUCT` | product name, brand, category, subcategory, unit price |
| `DIM_STORE` | store name, city, region, store format |
| `DIM_CUSTOMER` | customer name, segment, loyalty tier, city |

```
                    DIM_DATE
                        |
    DIM_STORE ---- FACT_SALES ---- DIM_PRODUCT
                        |
                   DIM_CUSTOMER
```

> The fact table holds only measures and keys, while every descriptive attribute
> lives in a dimension. That separation is the heart of the star schema.

**Note:** the ASCII sketch above shows the *shape*. Your deliverable is a proper
ER diagram — tables, key columns, and the lines between them.
